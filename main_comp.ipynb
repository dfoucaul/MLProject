```python

## Version 3 (Optimized/Corrected based on User Request)
# %% [markdown]
# # INFO-F-422 Project: EMG-based Hand Pose Estimation (Guided Gestures)
#
# Team: [Your Team Name/Number]
# Members: [Your Names]
#
# This notebook implements baseline and neural network approaches for predicting hand joint angles from sEMG signals using the guided gestures dataset.
# It includes feature extraction, cross-validation, baseline models (Ridge, Lasso, RF),
# feature selection (SFS), a sophisticated 3-layer Neural Network (NN_v3),
# and ensembling techniques (Averaging, Stacking).

# %%
# Core Libraries
import numpy as np
import pandas as pd  # Import pandas before seaborn
import matplotlib.pyplot as plt
import seaborn as sns  # Import seaborn after pandas
import time
import pickle # To save models/results
from contextlib import contextmanager
import gc # Garbage collection for potentially large objects/models

# Scikit-learn Modules
from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer # Example if needed, but unlikely for features
from sklearn.model_selection import GroupKFold, cross_validate, train_test_split, GridSearchCV, cross_val_predict
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.metrics import mean_squared_error, make_scorer, r2_score
from sklearn.linear_model import Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn import set_config # For metadata routing

# PyTorch Modules (for Objective 5 & Ensembling)
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
import torch.nn.functional as F

# SciPy (for filtering and potentially other signal processing)
import scipy.fftpack
from scipy.signal import butter, filtfilt

# Configure visualizations
%matplotlib inline
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Define a timing context manager for convenience
@contextmanager
def timer(label: str):
    start = time.time()
    print(f"{label}...")
    yield
    end = time.time()
    print(f"{label} done in {end - start:.2f} seconds.\n")

# Configure metadata routing (needed for SFS and potentially CV with params)
try:
    set_config(enable_metadata_routing=True)
    print(f"Scikit-learn metadata routing enabled: {set_config(True)}") # Confirm it's set
except Exception as e:
    print(f"Could not enable metadata routing (sklearn version might be older): {e}")

# Set device for PyTorch
# Leverage the powerful GPU specified (RTX 5070 Ti -> assume similar to 4070 Ti for now)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using PyTorch device: {device}")
if device == torch.device("cuda"):
    print(f"CUDA Device Name: {torch.cuda.get_device_name(0)}")

# Global constants
RANDOM_STATE = 42
N_JOBS = -1 # Use all available CPU cores for sklearn jobs

# %% [markdown]
# ## 1. Load Data
#
# Load the guided gestures dataset files.

# %%
# File paths
X_TRAIN_VAL_PATH = "guided_dataset_X.npy"
Y_TRAIN_VAL_PATH = "guided_dataset_y.npy"
X_TEST_PATH = "guided_testset_X.npy" # We only need shape info for now

with timer("Loading Data"):
    # Load the training/validation data
    X_train_val_raw = np.load(X_TRAIN_VAL_PATH)
    y_train_val_raw = np.load(Y_TRAIN_VAL_PATH)

    # Load the test data structure description (we need shape info)
    # Use mmap_mode='r' for memory efficiency if the file is large and we only need metadata
    X_test_raw_info = np.load(X_TEST_PATH, mmap_mode='r')

# Print shapes to verify
print("Raw Training/Validation Data Shapes:")
print(f"X_train_val_raw: {X_train_val_raw.shape}")
print(f"y_train_val_raw: {y_train_val_raw.shape}")
print("\nRaw Test Data Shape (for info):")
print(f"X_test_raw_info: {X_test_raw_info.shape}")

# Constants derived from data
N_SESSIONS_TRAIN_VAL = X_train_val_raw.shape[0]
N_ELECTRODES = X_train_val_raw.shape[1]
N_TIME_POINTS_TRAIN_VAL = X_train_val_raw.shape[2]
N_JOINTS = y_train_val_raw.shape[1] # Target dimension
FS = 1024 # Sampling frequency from documentation

print(f"\nNumber of Training/Validation Sessions: {N_SESSIONS_TRAIN_VAL}")
print(f"Number of Electrodes: {N_ELECTRODES}")
print(f"Number of Time Points per Session: {N_TIME_POINTS_TRAIN_VAL}")
print(f"Number of Joints (Targets): {N_JOINTS}")
print(f"Sampling Frequency (Hz): {FS}")


# %% [markdown]
# ## 2. Signal Filtering (Optional but Applied)
#
# Applying DC offset removal (mean subtraction per channel) and band-pass filtering (20-450 Hz) to the raw sEMG data before windowing. This helps remove noise and focus on the relevant frequency range for muscle activity.

# %%
def butter_bandpass(lowcut, highcut, fs, order=5):
     nyq = 0.5 * fs
     low = lowcut / nyq
     high = highcut / nyq
     # Catch potential issues with frequency limits
     if low <= 0 or high >= 1:
         raise ValueError(f"Butterworth filter frequencies invalid: low={low}, high={high} (must be between 0 and 1)")
     b, a = butter(order, [low, high], btype='band')
     return b, a

def bandpass_filter(data, lowcut, highcut, fs, order=5):
     """Applies a zero-phase bandpass filter to the data."""
     b, a = butter_bandpass(lowcut, highcut, fs, order=order)
     # Apply filter along the time axis (last axis)
     y = filtfilt(b, a, data, axis=-1)
     return y

LOWCUT = 20  # Hz
HIGHCUT = 450 # Hz
FILTER_ORDER = 4 # A slightly lower order can be more stable

with timer("Applying Signal Filtering"):
    X_train_val_filtered = np.zeros_like(X_train_val_raw)
    for i in range(N_SESSIONS_TRAIN_VAL):
        session_data = X_train_val_raw[i]
        # 1. Remove DC offset (mean) per channel along the time axis
        dc_removed = session_data - np.mean(session_data, axis=-1, keepdims=True)
        # 2. Apply bandpass filter
        X_train_val_filtered[i] = bandpass_filter(dc_removed, LOWCUT, HIGHCUT, FS, order=FILTER_ORDER)

    # Use the filtered data for the next step
    X_train_val_to_window = X_train_val_filtered
    print("Applied DC offset removal and bandpass filtering.")

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.plot(X_train_val_raw[0, 0, :5000])
plt.title("Raw Signal (Session 0, Electrode 0, First 5k points)")
plt.xlabel("Time points")
plt.ylabel("Amplitude")
plt.subplot(1, 2, 2)
plt.plot(X_train_val_filtered[0, 0, :5000])
plt.title("Filtered Signal (Session 0, Electrode 0, First 5k points)")
plt.xlabel("Time points")
plt.ylabel("Amplitude")
plt.tight_layout()
plt.show()

# Clean up raw data if memory is tight (optional)
# del X_train_val_raw
# gc.collect()


# %% [markdown]
# ## 3. Dataset Preparation: Overlapping Windows (Objective 2)
#
# Segment the filtered sEMG and corresponding raw joint angle data into overlapping windows. This creates the samples for our supervised learning task. Each window of EMG data will predict the joint angles at the *end* of that window.

# %%
# Windowing Parameters
WINDOW_SIZE = 500  # k (Number of time points per window, ~0.5 seconds)
OVERLAP_PERCENTAGE = 0.99 # 75% overlap -> step = 0.25 * window_size

step = int(WINDOW_SIZE * (1 - OVERLAP_PERCENTAGE))
if step <= 0:
    raise ValueError("Step size must be positive. Check window size and overlap percentage.")
print(f"Window Size (k): {WINDOW_SIZE}")
print(f"Overlap: {OVERLAP_PERCENTAGE*100:.0f}%")
print(f"Step Size: {step}")

def create_overlapping_windows(X_raw, y_raw, window_size, step):
    """
    Creates overlapping windows for EMG (X) and corresponding target poses (y).
    Also generates group labels (session index) for cross-validation.
    """
    n_sessions, n_electrodes, n_timepoints_total = X_raw.shape
    _, n_joints, _ = y_raw.shape # y_raw has shape (n_sessions, n_joints, n_timepoints_total)

    X_windowed_list = []
    y_windowed_list = []
    groups_list = []

    for session_idx in range(n_sessions):
        session_X = X_raw[session_idx] # Shape: (n_electrodes, n_timepoints_total)
        session_y = y_raw[session_idx] # Shape: (n_joints, n_timepoints_total)
        n_windows_session = 0

        # Calculate start indices for windows in this session
        start_indices = np.arange(0, n_timepoints_total - window_size + 1, step)

        for start_idx in start_indices:
            end_idx = start_idx + window_size
            window_X = session_X[:, start_idx:end_idx] # Shape: (n_electrodes, window_size)
            # Target is the pose at the *end* of the window
            target_y = session_y[:, end_idx - 1] # Shape: (n_joints,)

            X_windowed_list.append(window_X)
            y_windowed_list.append(target_y)
            groups_list.append(session_idx)
            n_windows_session += 1

        print(f"Session {session_idx}: Created {n_windows_session} windows.")

    # Stack the lists into numpy arrays
    X_windowed_np = np.stack(X_windowed_list) # Shape: (total_windows, n_electrodes, window_size)
    y_windowed_np = np.stack(y_windowed_list) # Shape: (total_windows, n_joints)
    groups_np = np.array(groups_list) # Shape: (total_windows,)

    return X_windowed_np, y_windowed_np, groups_np


# Create the windowed dataset using the filtered EMG data
with timer("Creating overlapping windows"):
    X_windowed, y_windowed, groups = create_overlapping_windows(
        X_train_val_to_window, y_train_val_raw, WINDOW_SIZE, step
    )

print("\nWindowed Data Shapes:")
print(f"X_windowed (Input Windows): {X_windowed.shape}")
print(f"y_windowed (Target Poses): {y_windowed.shape}")
print(f"Groups (Session IDs): {groups.shape}")
print(f"Unique groups (sessions): {np.unique(groups)}")

# Clean up intermediate large arrays if needed
# del X_train_val_to_window, X_train_val_filtered
# gc.collect()



# %% [markdown]
# ## 4. Data Preprocessing (Features)
#
# The primary preprocessing step applied *after* feature extraction is **standardization**. We use `StandardScaler` to remove the mean and scale features to unit variance. This is generally beneficial for linear models (Ridge, Lasso), distance-based algorithms, and Neural Networks.
#
# Standardization will be implemented *within* the Scikit-learn pipelines for baseline models and SFS, and applied explicitly *within each fold* of the cross-validation loop for the Neural Network to prevent data leakage from the validation set into the training set scaling parameters.

# %%
print("Preprocessing Step: StandardScaler will be applied to extracted features.")
print(" - Inside scikit-learn pipelines for baseline/SFS models.")
print(" - Within each CV fold for Neural Network training/validation.")

# %% [markdown]
# ## 5. Custom Feature Extractor (Baseline - Objective 4)
#
# Define a custom Scikit-learn transformer (`EMGFeatureExtractor`) to calculate time-domain and frequency-domain features from each EMG window. This includes the 6 required features (MAV, RMS, Var, STD, ZC, MPR) plus additional potentially useful ones (WL, SSC, IEMG, MNF, MDF).

# %%
class EMGFeatureExtractor(BaseEstimator, TransformerMixin):
    """
    Extracts time-domain and frequency-domain features from EMG windows.
    Handles potential NaNs during calculation.
    Allows selection of specific features to extract.
    """
    def __init__(self, zc_threshold=1e-5, ssc_threshold=1e-5, selected_features=None, fs=FS, verbose=False):
        # Store parameters passed to __init__
        self.zc_threshold = zc_threshold
        self.ssc_threshold = ssc_threshold
        self.selected_features = selected_features # Can be None (all), or a list of names
        self.fs = fs # Sampling frequency needed for freq-domain features
        self.verbose = verbose

        # Define all features this extractor *can* compute
        self.all_available_feature_names = sorted([
            'MAV', 'RMS', 'Var', 'STD', 'ZC', 'MPR',
            'WL', 'SSC', 'IEMG', 'MNF', 'MDF'
        ])
        self._actual_features_to_extract = [] # Will be set in fit()
        self.n_electrodes_ = None # Will be set in fit()

    def _resolve_features_to_extract(self):
        """Determine the actual list of features to use based on self.selected_features."""
        if self.selected_features is None:
            # Default to all available features if none specified
            return self.all_available_feature_names
        else:
            # Validate user-provided list
            current_selection = list(self.selected_features)
            validated_selection = []
            for f in current_selection:
                if f in self.all_available_feature_names:
                    validated_selection.append(f)
                else:
                    print(f"Warning: Ignoring invalid feature '{f}'. Allowed: {self.all_available_feature_names}")
            if not validated_selection:
                raise ValueError("No valid features selected.")
            # Return sorted unique list
            return sorted(list(set(validated_selection)))

    def fit(self, X, y=None):
        """Fits the extractor, mainly determining the features to extract and input shape."""
        # X shape is expected to be (n_windows, n_electrodes, window_size)
        if X.ndim != 3:
            raise ValueError(f"Input X must be 3D (n_windows, n_electrodes, window_size), got shape {X.shape}")
        self.n_electrodes_ = X.shape[1]
        self._actual_features_to_extract = self._resolve_features_to_extract()
        if self.verbose:
            print(f"Feature Extractor fit. Electrodes: {self.n_electrodes_}. Extracting features: {self._actual_features_to_extract}")
        return self

    def transform(self, X):
        """Transforms the raw EMG windows into a feature matrix."""
        # Ensure fit has been called
        if self.n_electrodes_ is None or not self._actual_features_to_extract:
             if self.verbose: print("Warning: fit() not called before transform(). Calling fit() now.")
             self.fit(X) # Call fit to set attributes

        n_windows, n_electrodes, window_size = X.shape
        if n_electrodes != self.n_electrodes_:
            raise ValueError(f"Input X has {n_electrodes} electrodes, but fitted on {self.n_electrodes_}")

        n_selected_features = len(self._actual_features_to_extract)
        if n_selected_features == 0:
             if self.verbose: print("Warning: No features selected to extract.")
             return np.empty((n_windows, 0))

        # Pre-allocate feature matrix: (n_windows, n_electrodes * n_selected_features)
        features_data = np.zeros((n_windows, self.n_electrodes_ * n_selected_features))

        # Pre-calculate FFT frequencies (same for all windows of the same size)
        freqs = np.fft.rfftfreq(window_size, d=1.0/self.fs)

        for i in range(n_windows):
            window_features_list = []
            for j in range(self.n_electrodes_):
                channel_data = X[i, j, :] # Data for window i, electrode j
                single_channel_features = {} # Store features for this channel here

                # --- Time Domain Features ---
                abs_channel_data = np.abs(channel_data)
                single_channel_features['MAV'] = np.mean(abs_channel_data)
                single_channel_features['RMS'] = np.sqrt(np.mean(channel_data**2))
                single_channel_features['Var'] = np.var(channel_data)
                # Handle potential negative variance due to floating point errors if signal is constant
                std_val = np.sqrt(max(0, single_channel_features['Var']))
                single_channel_features['STD'] = std_val
                # Zero Crossing: Count sign changes
                # Ensure comparison is robust to exact zero values if zc_threshold is 0
                adjusted_data = channel_data - self.zc_threshold
                sign_changes = np.diff(np.signbit(adjusted_data))
                single_channel_features['ZC'] = np.sum(sign_changes.astype(int) != 0)
                # Myopulse Percentage Rate: % samples exceeding a threshold (e.g., 1 standard deviation)
                mpr_thresh = std_val if std_val > 1e-9 else 1e-9 # Avoid division by zero/tiny std
                single_channel_features['MPR'] = np.mean(abs_channel_data > mpr_thresh)
                # Waveform Length
                single_channel_features['WL'] = np.sum(np.abs(np.diff(channel_data)))
                # Slope Sign Changes
                diff_data = np.diff(channel_data)
                ssc_cond1 = (diff_data[:-1] > self.ssc_threshold) & (diff_data[1:] < -self.ssc_threshold)
                ssc_cond2 = (diff_data[:-1] < -self.ssc_threshold) & (diff_data[1:] > self.ssc_threshold)
                single_channel_features['SSC'] = np.sum(ssc_cond1 | ssc_cond2)
                # Integrated EMG
                single_channel_features['IEMG'] = np.sum(abs_channel_data)

                # --- Frequency Domain Features ---
                fft_spectrum = np.abs(np.fft.rfft(channel_data))
                power_spectrum = fft_spectrum**2
                total_power = np.sum(power_spectrum)

                if total_power > 1e-9: # Avoid division by zero if signal is flat/zero
                    # Mean Frequency (MNF)
                    single_channel_features['MNF'] = np.sum(freqs * power_spectrum) / total_power
                    # Median Frequency (MDF)
                    cumulative_power = np.cumsum(power_spectrum)
                    mdf_index = np.searchsorted(cumulative_power, total_power / 2.0, side='left')
                    # Handle edge case where mdf_index might be out of bounds
                    if mdf_index >= len(freqs): mdf_index = len(freqs) - 1
                    single_channel_features['MDF'] = freqs[mdf_index]
                else:
                    single_channel_features['MNF'] = 0.0
                    single_channel_features['MDF'] = 0.0

                # Append only the selected features for this channel
                for feature_name in self._actual_features_to_extract:
                    window_features_list.append(single_channel_features[feature_name])

            # Assign the flattened features for this window
            features_data[i, :] = window_features_list

        # Handle potential NaNs/Infs that might arise from calculations (e.g., division by zero if not caught)
        return np.nan_to_num(features_data, nan=0.0, posinf=0.0, neginf=0.0)

    def get_feature_names_out(self, input_features=None):
        """Returns the names of the output features."""
        if self.n_electrodes_ is None or not self._actual_features_to_extract:
            # If called before fit, try to resolve based on init params
            temp_actual_features = self._resolve_features_to_extract()
            if self.verbose: print("Warning: get_feature_names_out() called on unfitted instance. Resolving features from init params.")
            # Assume a default number of electrodes if not fitted? Risky. Best practice is to fit first.
            # For robustness, let's return based on init params if available.
            temp_n_electrodes = self.n_electrodes_ if self.n_electrodes_ is not None else N_ELECTRODES # Use global if not fitted
        else:
            temp_actual_features = self._actual_features_to_extract
            temp_n_electrodes = self.n_electrodes_

        return [f"Elec{j+1}_{feat}" for j in range(temp_n_electrodes) for feat in temp_actual_features]


# --- Instantiate and Test the "Canonical" Feature Extractor ---
# This instance will extract ALL available features and will be used for the NN input
with timer("Testing Canonical Feature Extractor (all features)"):
    feature_extractor_canonical = EMGFeatureExtractor(fs=FS, verbose=True)
    # Fit the extractor on the windowed data
    feature_extractor_canonical.fit(X_windowed)
    # Transform the data to get the features
    X_features_canonical = feature_extractor_canonical.transform(X_windowed)

# This feature set is the input for the Neural Network
X_input_for_nn_cv = X_features_canonical
n_input_features_nn = X_input_for_nn_cv.shape[1]

print("\nCanonical Feature Matrix Shape (Input for NN):")
print(f"X_features_canonical: {X_features_canonical.shape}")

# Get the feature names generated by the canonical extractor
feature_names_canonical = feature_extractor_canonical.get_feature_names_out()
print(f"Generated {len(feature_names_canonical)} feature names.")
print(f"Example features: {feature_names_canonical[:5]}...{feature_names_canonical[-5:]}")

# Basic check for NaNs/Infs in the canonical features
if np.any(np.isnan(X_features_canonical)) or np.any(np.isinf(X_features_canonical)):
    print("\nWARNING: Found NaNs or Infs in the canonical feature matrix!")
else:
    print("\nCanonical feature matrix passed NaN/Inf check.")



# %% [markdown]
# ## 6. Cross-Validation Strategy (Objective 3)
#
# Use `GroupKFold` with session IDs as groups. This implements Leave-One-Session-Out Cross-Validation (LOSO CV), ensuring that data from the same recording session is kept together in either the training or validation set, preventing data leakage and providing a more realistic estimate of generalization performance to unseen sessions.

# %%
N_SPLITS = N_SESSIONS_TRAIN_VAL # = 5 for this dataset
cv_strategy = GroupKFold(n_splits=N_SPLITS)
print(f"Using GroupKFold with {N_SPLITS} splits (Leave-One-Session-Out).")

# Function to visualize CV splits (useful for understanding)
def plot_cv_indices(cv, X, y, group, n_splits, lw=10):
    """Creates a sample plot for training and validation indices"""
    fig, ax = plt.subplots(1, 1, figsize=(11, n_splits / 2)) # Adjust figsize
    cmap_data = plt.cm.Paired
    cmap_cv = plt.cm.coolwarm
    for i, (train_idx, test_idx) in enumerate(cv.split(X=X, y=y, groups=group)):
        indices = np.array([np.nan] * len(X))
        indices[test_idx] = 1
        indices[train_idx] = 0
        # Plot the split info
        ax.scatter(range(len(indices)), [i + .5] * len(indices),
                   c=indices, marker='_', lw=lw, cmap=cmap_cv,
                   vmin=-.2, vmax=1.2)
    # Plot the session groups
    ax.scatter(range(len(X)), [i + 1.5] * len(X),
               c=group, marker='_', lw=lw, cmap=cmap_data)
    # Formatting
    yticklabels = list(range(n_splits)) + ['Session']
    ax.set(yticks=np.arange(n_splits + 1) + .5, yticklabels=yticklabels,
           xlabel='Sample Index', ylabel="CV Fold / Session",
           ylim=[n_splits + 1.2, -.2], xlim=[0, len(y)])
    ax.set_title(f'{type(cv).__name__} CV Splits', fontsize=15)
    plt.tight_layout()
    return ax

# Visualize the splits using the canonical features (doesn't matter which features for visualization)
plot_cv_indices(cv_strategy, X_features_canonical, y_windowed, groups, N_SPLITS)
plt.show()




# %% [markdown]
# ## 7. Baseline Model Training and Evaluation (Objective 4 - Expanded)
#
# Evaluate three baseline regression models: Ridge, Lasso, and Random Forest.
# A Scikit-learn `Pipeline` is used to combine feature extraction (using the `EMGFeatureExtractor` with default settings - all features) and standardization (`StandardScaler`) before feeding the data to the regressor. Evaluation uses the LOSO CV strategy defined earlier.

# %%
# Define Performance Metrics (RMSE, NMSE, R2)
def root_mean_squared_error(y_true, y_pred):
    """Calculate Root Mean Squared Error."""
    return np.sqrt(mean_squared_error(y_true, y_pred))

def normalized_mean_squared_error(y_true, y_pred):
    """Calculate Normalized Mean Squared Error."""
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    if y_true.shape != y_pred.shape:
        raise ValueError(f"Shape mismatch: y_true {y_true.shape}, y_pred {y_pred.shape}")
    if y_true.ndim == 1:
        # Reshape if single target for consistency
        y_true = y_true.reshape(-1, 1)
        y_pred = y_pred.reshape(-1, 1)

    # Sum of squared errors (numerator)
    numerator = np.sum((y_true - y_pred)**2)
    # Variance of true values (denominator)
    mean_y_true_per_joint = np.mean(y_true, axis=0, keepdims=True) # Keep dims for broadcasting
    denominator = np.sum((y_true - mean_y_true_per_joint)**2)

    # Handle case where variance is zero (all true values are the same)
    if denominator < 1e-9:
        return 0.0 if numerator < 1e-9 else np.inf # NMSE is 0 if predictions are also constant, else infinite error
    else:
        return numerator / denominator

# Create scorer objects for use with scikit-learn's cross_validate
# Note: Scikit-learn maximises scores, so RMSE/NMSE need `greater_is_better=False`
scorers = {
    'rmse': make_scorer(root_mean_squared_error, greater_is_better=False),
    'nmse': make_scorer(normalized_mean_squared_error, greater_is_better=False),
    'r2': make_scorer(r2_score) # R2 is already maximized
}

# %%
# Select Baseline Models
model_ridge = Ridge(alpha=1.0, random_state=RANDOM_STATE) # Regularized linear model
model_lasso = Lasso(alpha=0.01, max_iter=3000, tol=1e-3, random_state=RANDOM_STATE) # L1 regularization, promotes sparsity
model_rf = RandomForestRegressor(
    n_estimators=100,    # Number of trees
    max_depth=15,        # Limit tree depth to prevent overfitting
    min_samples_split=10,# Min samples to split a node
    min_samples_leaf=5,  # Min samples in a leaf node
    max_features=0.33,   # Consider only 1/3 of features at each split
    n_jobs=N_JOBS,       # Use all CPU cores
    random_state=RANDOM_STATE
)

# Store baseline models in a dictionary for easy iteration
baseline_models_to_evaluate = {
    "Ridge": model_ridge,
    "Lasso": model_lasso,
    "RandomForest": model_rf
}

# Store all cross-validation results here across different objectives
all_cv_results = {}

# %%
# Evaluate baseline models using Cross-Validation
# %%
# Evaluate baseline models using Cross-Validation
print("--- Evaluating Baseline Models (Ridge, Lasso, RF) using LOSO CV ---")

# Make sure get_config is imported
from sklearn import get_config

for model_name, model in baseline_models_to_evaluate.items():
    with timer(f"Cross-validating {model_name}"):
        # Define the pipeline: Feature Extraction -> Scaling -> Regression
        pipeline = Pipeline([
            ('feature_extraction', EMGFeatureExtractor(fs=FS)), # Use canonical extractor (all features)
            ('scaling', StandardScaler()),
            ('regression', clone(model)) # Clone model for safety within CV
        ])

        # Perform cross-validation
        # Prepare base arguments
        cv_args = {
            'estimator': pipeline,
            'X': X_windowed,
            'y': y_windowed,
            'cv': cv_strategy,
            'scoring': scorers,
            'n_jobs': N_JOBS,
            'return_train_score': False,
            'error_score': 'raise' # See errors during CV
        }

        # Check if metadata routing is enabled using get_config()
        # --- CORRECTED LINE BELOW ---
        if get_config().get('enable_metadata_routing', False):
        # --- END OF CORRECTION ---
             cv_args['params'] = {'groups': groups}
             print(f"  (Using metadata routing for groups in {model_name})")
        else:
             # If metadata routing is not enabled, try passing groups directly
             # Note: Older sklearn versions might raise TypeError here if 'groups' is unexpected
             try:
                 cv_args['groups'] = groups
                 print(f"  (Passing groups directly for {model_name} - Metadata routing disabled)")
                 scores = cross_validate(**cv_args)
             except TypeError as e:
                 if 'got an unexpected keyword argument \'groups\'' in str(e):
                      print(f"  Warning: Passing 'groups' directly failed for {model_name}. Trying CV without explicit groups (may be incorrect if estimator needs groups).")
                      # Remove groups argument and proceed cautiously
                      del cv_args['groups']
                      scores = cross_validate(**cv_args)
                 else: raise e # Re-raise other TypeErrors
        # If the 'if' block executed successfully or the 'try' in the 'else' block succeeded
        if 'scores' not in locals() or scores is None: # Check if scores were assigned
             scores = cross_validate(**cv_args) # Run it (either with 'params' or without 'groups' if direct passing failed)


        all_cv_results[model_name] = scores # Store results
        scores = None # Reset scores for the next iteration's safety
# %%
# Display Baseline CV Results
print("\n--- Baseline Cross-Validation Performance Results ---")

results_summary_list = [] # List to store summary dictionaries for DataFrame

for model_name, scores in all_cv_results.items():
     if model_name in baseline_models_to_evaluate: # Ensure it's a baseline result
        # Note: RMSE/NMSE scorers return negative values because `greater_is_better=False`
        mean_rmse = -np.mean(scores['test_rmse'])
        std_rmse = np.std(scores['test_rmse']) # Std dev of the negative values, magnitude is correct
        mean_nmse = -np.mean(scores['test_nmse'])
        std_nmse = np.std(scores['test_nmse'])
        mean_r2 = np.mean(scores['test_r2'])
        std_r2 = np.std(scores['test_r2'])
        fit_time_mean = np.mean(scores['fit_time'])
        score_time_mean = np.mean(scores['score_time'])

        print(f"\n--- {model_name} ---")
        print(f"  Mean CV RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}")
        print(f"  Mean CV NMSE: {mean_nmse:.4f} +/- {std_nmse:.4f}")
        print(f"  Mean CV R2:   {mean_r2:.4f} +/- {std_r2:.4f}")
        print(f"  Avg Fit Time: {fit_time_mean:.2f}s")
        print(f"  Avg Score Time: {score_time_mean:.2f}s")

        # Append summary to list
        results_summary_list.append({
            'Model': model_name, 'Type': 'Baseline',
            'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse,
            'NMSE_mean': mean_nmse, 'NMSE_std': std_nmse,
            'R2_mean': mean_r2, 'R2_std': std_r2,
            'FitTime': fit_time_mean
        })

# Create a DataFrame for easier comparison later
results_df = pd.DataFrame(results_summary_list)
print("\n--- Baseline Summary Table ---")
# Display sorted by best RMSE
print(results_df.sort_values(by='RMSE_mean').to_string(index=False, float_format="%.4f"))

# %% [markdown]
# **Baseline Interpretation:**
# - Review the summary table. Which baseline model performed best in terms of mean CV RMSE?
# - Random Forest often performs well on tabular feature data like this.
# - Lasso's performance depends heavily on the `alpha` parameter; the current `alpha=0.01` might be suboptimal. A `LassoCV` could find a better alpha but takes longer.
# - Note the computation times. RF is typically slower to train than Ridge/Lasso.

# %% [markdown]
# ## 8. Feature Importance and Selection (Objective 4)

# %%
# Import additional libraries for optimization
from joblib import parallel_backend, Parallel, delayed
import psutil

# Configuration to optimize parallelization
N_CPUS = psutil.cpu_count(logical=False)  # Use physical cores count
MEMORY_LIMIT = None  # Auto memory limit
BACKEND = 'loky'  # Efficient joblib backend

# --- Feature Importances (from Random Forest) ---
print("\n--- Analyzing Feature Importance with Optimized Processing ---")

# Create optimized pipeline with efficient memory management
analysis_pipeline_fs = Pipeline([
    ('feature_extraction', EMGFeatureExtractor(fs=FS)),  # Canonical extractor
    ('scaling', StandardScaler())
], memory=MEMORY_LIMIT, verbose=False)

# Fit the pipeline and transform the full training data with optimized parallel processing
with timer("Extracting & Scaling features for Importance/SFS Analysis"):
    # Set appropriate backend for parallel processing
    with parallel_backend(BACKEND, n_jobs=N_JOBS):
        # Use X_windowed (raw EMG) as input to the pipeline
        X_features_scaled_for_analysis = analysis_pipeline_fs.fit_transform(X_windowed, y_windowed)
        # Get feature names AFTER fitting the pipeline
        feature_names_analysis = analysis_pipeline_fs.named_steps['feature_extraction'].get_feature_names_out()
        print(f"Shape of features for analysis: {X_features_scaled_for_analysis.shape}")
        print(f"Number of feature names for analysis: {len(feature_names_analysis)}")

# Train a final Random Forest model with optimized parameters
rf_params = {
    'n_jobs': N_JOBS,  # Utilize available CPU cores
    'verbose': 0,
    'warm_start': False,  # Better for single training
    'bootstrap': True,  # Improves parallel execution
    'oob_score': False,  # Skip OOB to save computing resources
}

with timer("Training RandomForest on full data for Importance Analysis"):
    # Use the same RF parameters as in baseline evaluation for consistency
    model_rf_final_for_importance = clone(model_rf)  # Clone baseline RF definition
    
    # Apply optimized parameters but keep original hyperparameters
    for param, value in rf_params.items():
        if hasattr(model_rf_final_for_importance, param):
            setattr(model_rf_final_for_importance, param, value)
    
    # Train RF with parallel backend for better CPU utilization
    with parallel_backend(BACKEND, n_jobs=N_JOBS):
        model_rf_final_for_importance.fit(X_features_scaled_for_analysis, y_windowed)

# Get and plot feature importances
importances = model_rf_final_for_importance.feature_importances_
indices = np.argsort(importances)[::-1]  # Sort features by importance (descending)

print("\nRandom Forest Feature Importances (Top 30):")
n_top_features = 30
importance_summary = []
for i in range(min(n_top_features, len(importances))):
    feature_index = indices[i]
    # Ensure feature_index is within bounds of feature_names_analysis
    if feature_index < len(feature_names_analysis):
        feature_name = feature_names_analysis[feature_index]
        importance_value = importances[feature_index]
        print(f"{i + 1}. Feature: {feature_name} ({importance_value:.4f})")
        importance_summary.append({'Feature': feature_name, 'Importance': importance_value})
    else:
        print(f"Warning: Index {feature_index} out of bounds for feature names list (length {len(feature_names_analysis)}).")

# Plotting the top N importances
plt.figure(figsize=(12, 8))
plt.title(f"Random Forest Feature Importances (Top {n_top_features})")
plt.bar(range(min(n_top_features, len(importances))), importances[indices[:n_top_features]], align='center')
# Ensure xticks match the plotted bars
plot_xticks = [feature_names_analysis[indices[i]] for i in range(min(n_top_features, len(importances)))]
plt.xticks(range(min(n_top_features, len(importances))), plot_xticks, rotation=90)
plt.xlim([-1, min(n_top_features, len(importances))])
plt.ylabel("Importance (Gini importance / Mean decrease in impurity)")
plt.tight_layout()
plt.show()

# %%
# --- Sequential Feature Selection (SFS) with Optimized Computation ---
print("\n--- Performing Sequential Feature Selection (SFS) using Ridge with Optimized Processing ---")

# Make sure get_config is imported
from sklearn import get_config

# Optimized SFS parameters
sfs = SequentialFeatureSelector(
    estimator=Ridge(alpha=1.0, random_state=RANDOM_STATE, solver='auto'),  # 'auto' solver for best performance
    n_features_to_select='auto',  # 'auto' selects based on CV score improvement plateau
    tol=5e-4,  # Stop if score improvement is less than this tolerance
    direction='forward',
    scoring=scorers['rmse'],  # Minimize RMSE
    cv=cv_strategy,  # Use the same LOSO CV strategy
    n_jobs=N_JOBS  # Utilize available CPU cores
)

# Prepare arguments for SFS fit
sfs_args = {
    'X': X_features_scaled_for_analysis,  # Use the scaled features created earlier
    'y': y_windowed,
}

# Check metadata routing and add groups accordingly
if get_config().get('enable_metadata_routing', False):
    sfs_args['groups'] = groups  # Pass via 'groups' key when routing enabled
    print("  (Using metadata routing for groups in SFS)")
else:
    print("  (Metadata routing disabled - relying on CV object for group splitting)")

# Fit SFS with optimized parallel processing
with timer("Fitting Sequential Feature Selector"):
    # Use optimized backend for parallel processing
    with parallel_backend(BACKEND, n_jobs=N_JOBS):
        # Pass the prepared arguments
        sfs.fit(**sfs_args)

# Get selected features information
selected_feature_mask = sfs.get_support()
selected_feature_indices = sfs.get_support(indices=True)
# Ensure indices are valid before accessing feature_names_analysis
valid_indices = [idx for idx in selected_feature_indices if idx < len(feature_names_analysis)]
selected_feature_names = [feature_names_analysis[i] for i in valid_indices]
n_selected_features = len(selected_feature_names)

print(f"\nSFS selected {n_selected_features} features out of {len(feature_names_analysis)}.")
if n_selected_features > 0:
    print("Selected Feature Names (first 50 shown):")
    print(selected_feature_names[:50])
    if n_selected_features > 50: print("...")

    # Store selected feature types for reuse in model building
    selected_feature_types_only = set()
    
    # Analyze the types and electrodes of selected features
    try:
        selected_types = [f.split('_', 1)[1] for f in selected_feature_names]  # Split only once
        selected_electrodes = [f.split('_')[0] for f in selected_feature_names]
        
        # Get unique feature types for later use
        selected_feature_types_only = set(selected_types)
        
        print("\nCounts of selected feature types:")
        print(pd.Series(selected_types).value_counts())
        print("\nCounts of selected electrodes:")
        print(pd.Series(selected_electrodes).value_counts())
    except IndexError:
        print("\nWarning: Could not parse feature types/electrodes from SFS names.")
        print("Selected names:", selected_feature_names)
else:
    print("\nWarning: SFS selected 0 features. Check SFS parameters (e.g., tol) or input data.")

# Clean up to free memory
if 'model_rf_final_for_importance' in locals():
    del model_rf_final_for_importance

# Explicitly free memory if available
if torch.cuda.is_available():
    torch.cuda.empty_cache()
gc.collect()

# %% [markdown]
# ## 9. Final Baseline Pipeline with Selected Features (Optional Refinement)
#
# Evaluate the *best-performing baseline model* (identified from Section 7) using *only* the feature *types* selected by SFS. This tests if the feature selection simplified the model without significantly hurting (or maybe even improving) performance.

# %%
# Ensure get_config is imported (should be done at the top)
from sklearn import get_config

# Variable to store the feature types derived from SFS names
selected_feature_types_only = []
if 'n_selected_features' in locals() and n_selected_features > 0: # Check if SFS ran and selected features
     try:
        # Extract unique feature types from the selected names
        selected_feature_types_only = sorted(list(set(f.split('_', 1)[1] for f in selected_feature_names)))
     except IndexError:
        print("Warning: Could not extract feature types from SFS names. Skipping SFS pipeline evaluation.")
        n_selected_features = 0 # Treat as if no features selected or parsing failed
else:
     n_selected_features = 0 # Ensure it's 0 if SFS didn't run or selected none

# Proceed only if SFS selected a subset of features
if n_selected_features > 0 and 'feature_names_analysis' in locals() and n_selected_features < len(feature_names_analysis):
    print("\n--- Evaluating Best Baseline Model with SFS-Selected Feature Types ---")
    # Determine the best model from initial baseline CV (based on lowest mean RMSE)
    if 'results_df' in locals() and not results_df.empty and 'RMSE_mean' in results_df.columns:
        # Filter to only baseline models for this selection
        baseline_results = results_df[results_df['Type'] == 'Baseline']
        if not baseline_results.empty:
            best_baseline_model_name = baseline_results.loc[baseline_results['RMSE_mean'].idxmin()]['Model']
            print(f"Identified best baseline model from CV: {best_baseline_model_name}")

            # Get the corresponding model instance definition
            if best_baseline_model_name in baseline_models_to_evaluate:
                 best_model_instance_sfs = clone(baseline_models_to_evaluate[best_baseline_model_name])
            else:
                 print(f"Warning: Best baseline model '{best_baseline_model_name}' not found in dictionary. Cannot proceed.")
                 best_model_instance_sfs = None # Mark as None

            if best_model_instance_sfs is not None:
                print(f"Using {len(selected_feature_types_only)} feature types selected by SFS: {selected_feature_types_only}")

                # Create the SFS pipeline: Extract only selected features -> Scale -> Regress
                sfs_pipeline = Pipeline([
                    ('feature_extraction', EMGFeatureExtractor(fs=FS, selected_features=selected_feature_types_only)),
                    ('scaling', StandardScaler()),
                    ('regression', best_model_instance_sfs) # Use the cloned best baseline regressor
                ])

                model_name_sfs = f"{best_baseline_model_name}_SFS"
                print(f"\nEvaluating pipeline: {model_name_sfs}")
                with timer(f"Cross-validating {model_name_sfs}"):
                    # Prepare base arguments for cross_validate
                    sfs_cv_args = {
                        'estimator': sfs_pipeline, 'X': X_windowed, 'y': y_windowed,
                        'cv': cv_strategy, 'scoring': scorers, 'n_jobs': N_JOBS,
                        'return_train_score': False, 'error_score': 'raise'
                    }

                    sfs_scores = None # Initialize scores variable

                    # Check metadata routing and add groups
                    # --- CORRECTED LINE BELOW ---
                    if get_config().get('enable_metadata_routing', False):
                    # --- END OF CORRECTION ---
                        sfs_cv_args['params'] = {'groups': groups}
                        print(f"  (Using metadata routing for groups in {model_name_sfs} CV)")
                    else:
                        # Try passing groups directly if metadata routing is off
                        try:
                            sfs_cv_args['groups'] = groups
                            print(f"  (Passing groups directly for {model_name_sfs} CV - Metadata routing disabled)")
                            sfs_scores = cross_validate(**sfs_cv_args)
                        except TypeError as e:
                            if 'got an unexpected keyword argument \'groups\'' in str(e):
                                print(f"  Warning: Passing 'groups' directly failed for {model_name_sfs}. Trying CV without explicit groups.")
                                del sfs_cv_args['groups'] # Remove groups argument
                                sfs_scores = cross_validate(**sfs_cv_args) # Try again without groups
                            else: raise e # Re-raise other TypeErrors

                    # If the 'if' block succeeded or the first 'try'/'except' in the 'else' block succeeded
                    if sfs_scores is None:
                         sfs_scores = cross_validate(**sfs_cv_args) # Run it (either with params or corrected args)

                    # Ensure scores were obtained before proceeding
                    if sfs_scores is None:
                         raise RuntimeError(f"Cross-validation failed to produce scores for {model_name_sfs}.")

                    all_cv_results[model_name_sfs] = sfs_scores # Store results

                # Add SFS results to summary list
                mean_rmse = -np.mean(sfs_scores['test_rmse']); std_rmse = np.std(sfs_scores['test_rmse'])
                mean_nmse = -np.mean(sfs_scores['test_nmse']); std_nmse = np.std(sfs_scores['test_nmse'])
                mean_r2 = np.mean(sfs_scores['test_r2']); std_r2 = np.std(sfs_scores['test_r2'])
                fit_time_mean = np.mean(sfs_scores['fit_time'])

                print(f"\n--- {model_name_sfs} ---")
                print(f"  Mean CV RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}")
                print(f"  Mean CV NMSE: {mean_nmse:.4f} +/- {std_nmse:.4f}")
                print(f"  Mean CV R2:   {mean_r2:.4f} +/- {std_r2:.4f}")

                # Append summary results
                results_summary_list.append({
                    'Model': model_name_sfs, 'Type': 'Baseline_SFS',
                    'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse, 'NMSE_mean': mean_nmse,
                    'NMSE_std': std_nmse, 'R2_mean': mean_r2, 'R2_std': std_r2,
                    'FitTime': fit_time_mean
                })
                # Update results_df
                results_df = pd.DataFrame(results_summary_list)
                print("\n--- Updated Summary Table ---")
                print(results_df.sort_values(by='RMSE_mean').to_string(index=False, float_format="%.4f"))

        else: # If baseline_results is empty
            print("Skipping SFS pipeline evaluation as no baseline results were found in results_df.")
    else: # If results_df is empty or missing columns
        print("Skipping SFS pipeline evaluation as baseline results DataFrame is missing or malformed.")
else: # If SFS selected 0 or all features, or parsing failed
    print("\nSkipping evaluation with SFS features (SFS selected 0 or all features, or feature types could not be parsed).")


# %% [markdown]
# ## 10. Sophisticated Approach: Neural Network (NN_v3) (Objective 5)

# %%
# Add necessary imports for optimizations
from torch.cuda.amp import autocast, GradScaler
import torch.cuda.amp

# Set CUDA optimization flags to maximize GPU performance
torch.backends.cudnn.benchmark = True
torch.backends.cuda.matmul.allow_tf32 = True  # For Ampere+ GPUs (like your RTX 5070 Ti)
torch.backends.cudnn.allow_tf32 = True
torch.set_float32_matmul_precision('high')

# Data prefetcher for asynchronous data loading
class DataPrefetcher:
    def __init__(self, loader, device):
        self.loader = iter(loader)
        self.device = device
        self.stream = torch.cuda.Stream()
        self.preload()
        
    def preload(self):
        try:
            self.next_input, self.next_target = next(self.loader)
        except StopIteration:
            self.next_input = None
            self.next_target = None
            return
        
        with torch.cuda.stream(self.stream):
            self.next_input = self.next_input.to(self.device, non_blocking=True).float()
            self.next_target = self.next_target.to(self.device, non_blocking=True).float()
            
    def next(self):
        torch.cuda.current_stream().wait_stream(self.stream)
        input = self.next_input
        target = self.next_target
        self.preload()
        return input, target

# ---------- Helper Functions for PyTorch Training/Evaluation ----------
def train_nn_epoch(model, loader, criterion, optimizer, device):
    """Trains the model for one epoch."""
    model.train()  # Set model to training mode
    running_loss = 0.0
    
    # Use GradScaler for mixed precision training
    scaler = GradScaler()
    
    # Use prefetcher for faster data loading
    prefetcher = DataPrefetcher(loader, device)
    inputs, targets = prefetcher.next()
    batch_idx = 0
    
    while inputs is not None:
        batch_idx += 1
        
        # Zero the parameter gradients
        optimizer.zero_grad(set_to_none=True)  # Faster than zero_grad()
        
        # Mixed precision forward pass
        with autocast():
            outputs = model(inputs)
            loss = criterion(outputs, targets)
        
        # Scaled backward pass
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        
        running_loss += loss.item() * inputs.size(0)
        inputs, targets = prefetcher.next()
    
    epoch_loss = running_loss / len(loader.dataset)
    return epoch_loss

def evaluate_nn(model, loader, criterion, device):
    """Evaluates the model on a dataset."""
    model.eval()  # Set model to evaluation mode
    running_loss = 0.0
    all_preds = []
    all_targets = []
    
    # Use prefetcher for faster data loading during evaluation too
    prefetcher = DataPrefetcher(loader, device)
    inputs, targets = prefetcher.next()
    
    with torch.no_grad():  # Disable gradient calculation for evaluation
        while inputs is not None:
            # Mixed precision forward pass (for consistency with training)
            with autocast():
                outputs = model(inputs)
                loss = criterion(outputs, targets)
            
            running_loss += loss.item() * inputs.size(0)
            
            # Store predictions and targets for metric calculation outside loop
            all_preds.append(outputs.cpu().numpy())
            all_targets.append(targets.cpu().numpy())
            
            inputs, targets = prefetcher.next()
    
    epoch_loss = running_loss / len(loader.dataset)  # Average loss per sample
    # Concatenate predictions and targets from all batches
    all_preds_np = np.concatenate(all_preds, axis=0)
    all_targets_np = np.concatenate(all_targets, axis=0)
    
    return epoch_loss, all_preds_np, all_targets_np

# ---------- Early-Stopping Callback ----------
# EarlyStop class remains unchanged
class EarlyStop:
    """Simple early stopping implementation."""
    def __init__(self, patience=10, min_delta=1e-4):
        self.patience = patience
        self.min_delta = min_delta  # Minimum change to qualify as improvement
        self.counter = 0
        self.best_loss = float('inf')
        self.early_stop = False

    def __call__(self, val_loss):
        if val_loss < self.best_loss - self.min_delta:
            self.best_loss = val_loss
            self.counter = 0  # Reset counter on improvement
        else:
            self.counter += 1
            if self.counter >= self.patience:
                print(f"EarlyStopping: Patience ({self.patience}) reached. Stopping.")
                self.early_stop = True
        return self.early_stop

# ---------- Neural Network Architecture (NN_v3) ----------
# EMG_NN class definition remains the same
class EMG_NN(nn.Module):
    """3-Hidden Layer Feed-Forward Neural Network with BatchNorm and Dropout."""
    def __init__(self, n_features, n_joints, hidden_sizes=(256, 128, 64), dropout_p=0.2):
        super().__init__()
        h1, h2, h3 = hidden_sizes

        self.input_layer = nn.Linear(n_features, h1)
        self.bn1 = nn.BatchNorm1d(h1)

        self.hidden2 = nn.Linear(h1, h2)
        self.bn2 = nn.BatchNorm1d(h2)

        self.hidden3 = nn.Linear(h2, h3)
        self.bn3 = nn.BatchNorm1d(h3)

        self.output_layer = nn.Linear(h3, n_joints)

        self.dropout = nn.Dropout(dropout_p)

    def forward(self, x):
        # Layer 1
        x = self.input_layer(x)
        x = self.bn1(x)
        x = F.relu(x)
        x = self.dropout(x)
        # Layer 2
        x = self.hidden2(x)
        x = self.bn2(x)
        x = F.relu(x)
        x = self.dropout(x)
        # Layer 3
        x = self.hidden3(x)
        x = self.bn3(x)
        x = F.relu(x)
        x = self.dropout(x)
        # Output Layer
        x = self.output_layer(x)
        return x

# ---------- Hyperparameters for NN_v3 Training ----------
MAX_EPOCHS_NN_V3   = 150
BATCH_SIZE_NN_V3   = 1024
LR_NN_V3           = 1e-3
WEIGHT_DECAY_NN_V3 = 5e-5
PATIENCE_LR_NN_V3  = 12
PATIENCE_ES_NN_V3  = 20
NUM_WORKERS_LOADER = 8

# Create optimized DataLoader creation function (reused throughout the code)
def create_optimized_dataloader(dataset, batch_size, shuffle=True, num_workers=NUM_WORKERS_LOADER, 
                               pin_memory=True if torch.cuda.is_available() else False):
    return DataLoader(
        dataset, 
        batch_size=batch_size, 
        shuffle=shuffle, 
        num_workers=num_workers, 
        pin_memory=pin_memory,
        persistent_workers=True if num_workers > 0 else False,  # Keep workers alive between iterations
        prefetch_factor=2 if num_workers > 0 else None  # Prefetch 2 batches per worker
    )

# ---------- Mini-Grid Search (Quick Hyperparameter Check) ----------
print("--- NN-v3: Mini-Grid Hyperparameter Search (Quick Check) ---")
mini_grid = [
    {"hidden": (128, 64, 32), "dropout": 0.25},
    {"hidden": (256, 128, 64), "dropout": 0.2},
    {"hidden": (512, 256, 128), "dropout": 0.3},
]
best_cfg_nn_v3, best_rmse_nn_v3_grid = None, float('inf')

if 'X_input_for_nn_cv' not in locals() or X_input_for_nn_cv.shape[1] != len(feature_names_canonical):
     raise ValueError("Canonical features 'X_input_for_nn_cv' not found or shape mismatch. Run Section 5.")

try:
    Xtr_nn_grid, Xva_nn_grid, ytr_nn_grid, yva_nn_grid = train_test_split(
        X_input_for_nn_cv, y_windowed, test_size=0.20, random_state=RANDOM_STATE, stratify=groups
    )
except ValueError as e:
     print(f"Warning: Stratification failed in train_test_split ({e}). Splitting without stratification.")
     Xtr_nn_grid, Xva_nn_grid, ytr_nn_grid, yva_nn_grid = train_test_split(
        X_input_for_nn_cv, y_windowed, test_size=0.20, random_state=RANDOM_STATE
    )

for cfg_idx, cfg in enumerate(mini_grid):
    print(f"\nTesting NN-v3 grid config {cfg_idx+1}/{len(mini_grid)}: {cfg}")
    sc_nn_grid = StandardScaler()
    Xtr_nn_grid_scaled = sc_nn_grid.fit_transform(Xtr_nn_grid)
    Xva_nn_grid_scaled = sc_nn_grid.transform(Xva_nn_grid)

    # Create optimized tensor datasets
    tr_ds_grid = TensorDataset(torch.tensor(Xtr_nn_grid_scaled, dtype=torch.float32), 
                               torch.tensor(ytr_nn_grid, dtype=torch.float32))
    va_ds_grid = TensorDataset(torch.tensor(Xva_nn_grid_scaled, dtype=torch.float32), 
                               torch.tensor(yva_nn_grid, dtype=torch.float32))
    
    # Create optimized DataLoaders
    tr_ld_grid = create_optimized_dataloader(tr_ds_grid, batch_size=BATCH_SIZE_NN_V3, shuffle=True)
    va_ld_grid = create_optimized_dataloader(va_ds_grid, batch_size=BATCH_SIZE_NN_V3 * 2, shuffle=False)

    model_nn_grid = EMG_NN(n_input_features_nn, N_JOINTS, hidden_sizes=cfg['hidden'], dropout_p=cfg['dropout']).to(device)
    model_nn_grid.float()
    optimizer_nn_grid = torch.optim.Adam(model_nn_grid.parameters(), lr=LR_NN_V3, weight_decay=WEIGHT_DECAY_NN_V3)
    scheduler_nn_grid = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer_nn_grid, mode='min', factor=0.5, patience=PATIENCE_LR_NN_V3 // 2, verbose=False)
    early_stopper_grid = EarlyStop(patience=PATIENCE_ES_NN_V3 // 2)

    for epoch_grid in range(MAX_EPOCHS_NN_V3 // 2):
        train_loss_grid = train_nn_epoch(model_nn_grid, tr_ld_grid, nn.MSELoss(), optimizer_nn_grid, device)
        mse_val_grid, _, _ = evaluate_nn(model_nn_grid, va_ld_grid, nn.MSELoss(), device)
        rmse_val_grid = np.sqrt(mse_val_grid)
        scheduler_nn_grid.step(mse_val_grid)
        if early_stopper_grid(mse_val_grid):
            print(f"  Grid Early stopping at epoch {epoch_grid+1}. Val RMSE: {rmse_val_grid:.4f}")
            break

    final_mse_val_grid, _, _ = evaluate_nn(model_nn_grid, va_ld_grid, nn.MSELoss(), device)
    final_rmse_val_grid = np.sqrt(final_mse_val_grid)
    print(f"  Config {cfg} -> Final Validation RMSE: {final_rmse_val_grid:.4f}")

    if final_rmse_val_grid < best_rmse_nn_v3_grid:
        best_rmse_nn_v3_grid = final_rmse_val_grid
        best_cfg_nn_v3 = cfg

    del model_nn_grid, optimizer_nn_grid, scheduler_nn_grid, early_stopper_grid, tr_ld_grid, va_ld_grid
    if device == torch.device('cuda'): 
        torch.cuda.empty_cache()
    gc.collect()

print(f"\n✅ NN-v3 Best Config from Mini-Grid: {best_cfg_nn_v3} (Validation RMSE: {best_rmse_nn_v3_grid:.4f})")
if best_cfg_nn_v3 is None:
    print("Using default/user-provided config as grid search failed or yielded no results.")
    best_cfg_nn_v3 = {'hidden': (256, 128, 64), 'dropout': 0.2}

# ---------- Cross-Validation (LOSO) for NN_v3 with Best Config ----------
print(f"\n--- NN-v3: Cross-Validation ({N_SPLITS}-Fold LOSO) with Best Config ---")
print(f"Using Config: {best_cfg_nn_v3}")

fold_rmses_nn_v3 = []
fold_nmses_nn_v3 = []
fold_r2s_nn_v3 = []
fold_fit_times_nn_v3 = []

for k_fold, (train_idx, val_idx) in enumerate(cv_strategy.split(X_input_for_nn_cv, y_windowed, groups=groups)):
    fold_start_time = time.time()
    print(f"\nProcessing NN-v3 Fold {k_fold+1}/{N_SPLITS}")

    Xtr_fold_nn, Xva_fold_nn = X_input_for_nn_cv[train_idx], X_input_for_nn_cv[val_idx]
    ytr_fold_nn, yva_fold_nn = y_windowed[train_idx], y_windowed[val_idx]

    sc_nn_fold = StandardScaler()
    Xtr_fold_nn_scaled = sc_nn_fold.fit_transform(Xtr_fold_nn)
    Xva_fold_nn_scaled = sc_nn_fold.transform(Xva_fold_nn)

    # Create optimized tensor datasets
    tr_ds_fold = TensorDataset(torch.tensor(Xtr_fold_nn_scaled, dtype=torch.float32), 
                               torch.tensor(ytr_fold_nn, dtype=torch.float32))
    va_ds_fold = TensorDataset(torch.tensor(Xva_fold_nn_scaled, dtype=torch.float32), 
                               torch.tensor(yva_fold_nn, dtype=torch.float32))
    
    # Create optimized DataLoaders
    tr_ld_fold = create_optimized_dataloader(tr_ds_fold, batch_size=BATCH_SIZE_NN_V3, shuffle=True)
    va_ld_fold = create_optimized_dataloader(va_ds_fold, batch_size=BATCH_SIZE_NN_V3 * 2, shuffle=False)

    model_fold_nn_v3 = EMG_NN(n_input_features_nn, N_JOINTS,
                              hidden_sizes=best_cfg_nn_v3['hidden'], dropout_p=best_cfg_nn_v3['dropout']).to(device)
    model_fold_nn_v3.float()
    optimizer_fold_nn_v3 = torch.optim.Adam(model_fold_nn_v3.parameters(), lr=LR_NN_V3, weight_decay=WEIGHT_DECAY_NN_V3)
    scheduler_fold_nn_v3 = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer_fold_nn_v3, mode='min', factor=0.5, patience=PATIENCE_LR_NN_V3, verbose=False)
    early_stopper_fold_nn_v3 = EarlyStop(patience=PATIENCE_ES_NN_V3)

    print(f"  Training Fold {k_fold+1}...")
    best_fold_epoch = -1
    best_fold_val_loss = float('inf')

    for epoch_fold in range(MAX_EPOCHS_NN_V3):
        train_loss_fold = train_nn_epoch(model_fold_nn_v3, tr_ld_fold, nn.MSELoss(), optimizer_fold_nn_v3, device)
        mse_val_fold, _, _ = evaluate_nn(model_fold_nn_v3, va_ld_fold, nn.MSELoss(), device)
        
        scheduler_fold_nn_v3.step(mse_val_fold)
        if early_stopper_fold_nn_v3(mse_val_fold):
             best_fold_val_loss = early_stopper_fold_nn_v3.best_loss
             best_fold_epoch = epoch_fold + 1 - PATIENCE_ES_NN_V3
             print(f"    Fold {k_fold+1} early stopping triggered at epoch {epoch_fold+1}. Best Validation MSE occurred around epoch {best_fold_epoch} (Loss: {best_fold_val_loss:.4f}).")
             break
        elif mse_val_fold < best_fold_val_loss:
             best_fold_val_loss = mse_val_fold
             best_fold_epoch = epoch_fold + 1

    final_mse_fold, y_pred_fold_nn_v3, y_true_fold_nn_v3 = evaluate_nn(model_fold_nn_v3, va_ld_fold, nn.MSELoss(), device)
    fold_rmse = np.sqrt(final_mse_fold)
    fold_nmse = normalized_mean_squared_error(y_true_fold_nn_v3, y_pred_fold_nn_v3)
    fold_r2 = r2_score(y_true_fold_nn_v3, y_pred_fold_nn_v3)
    fold_end_time = time.time()
    fold_fit_time = fold_end_time - fold_start_time

    fold_rmses_nn_v3.append(fold_rmse)
    fold_nmses_nn_v3.append(fold_nmse)
    fold_r2s_nn_v3.append(fold_r2)
    fold_fit_times_nn_v3.append(fold_fit_time)
    print(f"  Fold {k_fold+1} finished. Time: {fold_fit_time:.2f}s")
    print(f"    Val RMSE: {fold_rmse:.4f}, Val NMSE: {fold_nmse:.4f}, Val R2: {fold_r2:.4f}")

    del model_fold_nn_v3, optimizer_fold_nn_v3, scheduler_fold_nn_v3, early_stopper_fold_nn_v3
    del tr_ds_fold, va_ds_fold, tr_ld_fold, va_ld_fold
    if device == torch.device('cuda'): 
        torch.cuda.empty_cache()
    gc.collect()

mean_cv_rmse_nn_v3 = np.mean(fold_rmses_nn_v3); std_cv_rmse_nn_v3 = np.std(fold_rmses_nn_v3)
mean_cv_nmse_nn_v3 = np.mean(fold_nmses_nn_v3); std_cv_nmse_nn_v3 = np.std(fold_nmses_nn_v3)
mean_cv_r2_nn_v3 = np.mean(fold_r2s_nn_v3); std_cv_r2_nn_v3 = np.std(fold_r2s_nn_v3)
mean_fit_time_nn_v3 = np.mean(fold_fit_times_nn_v3)

print(f"\n--- NN_v3 Cross-Validation Summary ---")
print(f"  Mean CV RMSE: {mean_cv_rmse_nn_v3:.4f} +/- {std_cv_rmse_nn_v3:.4f}")
print(f"  Mean CV NMSE: {mean_cv_nmse_nn_v3:.4f} +/- {std_cv_nmse_nn_v3:.4f}")
print(f"  Mean CV R2:   {mean_cv_r2_nn_v3:.4f} +/- {std_cv_r2_nn_v3:.4f}")
print(f"  Avg Fit Time per Fold: {mean_fit_time_nn_v3:.2f}s")

all_cv_results['NeuralNetwork_v3'] = {
    'test_rmse': np.array([-r for r in fold_rmses_nn_v3]), 'test_nmse': np.array([-n for n in fold_nmses_nn_v3]),
    'test_r2': np.array(fold_r2s_nn_v3), 'fit_time': np.array(fold_fit_times_nn_v3),
    'score_time': np.array([np.nan] * N_SPLITS)
}
results_summary_list.append({
    'Model': 'NeuralNetwork_v3', 'Type': 'Sophisticated',
    'RMSE_mean': mean_cv_rmse_nn_v3, 'RMSE_std': std_cv_rmse_nn_v3,
    'NMSE_mean': mean_cv_nmse_nn_v3, 'NMSE_std': std_cv_nmse_nn_v3,
    'R2_mean': mean_cv_r2_nn_v3, 'R2_std': std_cv_r2_nn_v3,
    'FitTime': mean_fit_time_nn_v3
})
results_df = pd.DataFrame(results_summary_list)
print("\n--- Updated Summary Table ---")
print(results_df.sort_values(by='RMSE_mean').to_string(index=False, float_format="%.4f"))

# ---------- Final Training of NN_v3 on 100% Training Data ----------
print("\n--- NN-v3: Final Training on All Training Data & Saving ---")
sc_final_nn_v3 = StandardScaler()
X_all_nn_v3_scaled = sc_final_nn_v3.fit_transform(X_input_for_nn_cv)

# Create optimized tensor dataset
all_dataset_nn_v3 = TensorDataset(torch.tensor(X_all_nn_v3_scaled, dtype=torch.float32), 
                                 torch.tensor(y_windowed, dtype=torch.float32))

# Create optimized DataLoader
all_loader_nn_v3 = create_optimized_dataloader(all_dataset_nn_v3, batch_size=BATCH_SIZE_NN_V3, shuffle=True)

final_model_nn_v3 = EMG_NN(n_input_features_nn, N_JOINTS,
                           hidden_sizes=best_cfg_nn_v3['hidden'], dropout_p=best_cfg_nn_v3['dropout']).to(device)
final_model_nn_v3.float()
optimizer_final_nn_v3 = torch.optim.Adam(final_model_nn_v3.parameters(), lr=LR_NN_V3, weight_decay=WEIGHT_DECAY_NN_V3)

final_epochs_nn_v3 = MAX_EPOCHS_NN_V3
print(f"Training final NN-v3 model for {final_epochs_nn_v3} epochs...")
with timer(f"Final NN-v3 training"):
    for epoch_final in range(final_epochs_nn_v3):
        final_train_loss = train_nn_epoch(final_model_nn_v3, all_loader_nn_v3, nn.MSELoss(), optimizer_final_nn_v3, device)
        if (epoch_final + 1) % 10 == 0 or epoch_final == final_epochs_nn_v3 - 1:
            print(f"  Final Training Epoch {epoch_final+1}/{final_epochs_nn_v3}, Training Loss: {final_train_loss:.6f}")

MODEL_NN_V3_PATH = "final_NN_v3_model.pth"
SCALER_NN_V3_PATH = "final_NN_v3_scaler.pkl"
torch.save(final_model_nn_v3.state_dict(), MODEL_NN_V3_PATH)
with open(SCALER_NN_V3_PATH, "wb") as f:
    pickle.dump(sc_final_nn_v3, f)
print(f"💾 NN-v3: Final model saved to {MODEL_NN_V3_PATH}")
print(f"💾 NN-v3: Final scaler saved to {SCALER_NN_V3_PATH}")

final_trained_model_dict_nn_v3 = {
    'model_type': 'NeuralNetwork_v3',
    'model': final_model_nn_v3,
    'scaler': sc_final_nn_v3,
    'feature_extractor': EMGFeatureExtractor(fs=FS)
}
# Add this to confirm GPU usage
print(f"Using device: {device}")
print(f"GPU available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU name: {torch.cuda.get_device_name(0)}")
    print(f"Current device: {torch.cuda.current_device()}")
# %% [markdown]
# ## 11. Objective 6: Ensembling Strategies

# First add the necessary imports for our optimized functions
from torch.cuda.amp import autocast, GradScaler

# Keep the DataPrefetcher class from previous optimization
class DataPrefetcher:
    def __init__(self, loader, device):
        self.loader = iter(loader)
        self.device = device
        self.stream = torch.cuda.Stream()
        self.preload()
        
    def preload(self):
        try:
            self.next_input, self.next_target = next(self.loader)
        except StopIteration:
            self.next_input = None
            self.next_target = None
            return
        
        with torch.cuda.stream(self.stream):
            self.next_input = self.next_input.to(self.device, non_blocking=True).float()
            self.next_target = self.next_target.to(self.device, non_blocking=True).float()
            
    def next(self):
        torch.cuda.current_stream().wait_stream(self.stream)
        input = self.next_input
        target = self.next_target
        self.preload()
        return input, target

# Create optimized DataLoader creation function
def create_optimized_dataloader(dataset, batch_size, shuffle=True, num_workers=NUM_WORKERS_LOADER, 
                               pin_memory=True if torch.cuda.is_available() else False):
    return DataLoader(
        dataset, 
        batch_size=batch_size, 
        shuffle=shuffle, 
        num_workers=num_workers, 
        pin_memory=pin_memory,
        persistent_workers=True if num_workers > 0 else False,  # Keep workers alive between iterations
        prefetch_factor=2 if num_workers > 0 else None  # Prefetch 2 batches per worker
    )

# %% [markdown]
# ### 11.1 Generate Out-of-Fold (OOF) Predictions

# %%
oof_predictions = {} # Dictionary to store OOF predictions arrays

# --- OOF Predictions for Scikit-learn Baseline Models ---
print("--- Generating OOF Predictions for Baseline Models (Ridge, Lasso, RF) ---")
# Use baseline_models_to_evaluate dict (Ridge, Lasso, RF instances)

# Ensure get_config is imported (should be done at the top)
from sklearn import get_config

for model_name, model_instance in baseline_models_to_evaluate.items():
     # Pipeline for OOF: Extract all features -> Scale -> Regress
     pipeline_oof_base = Pipeline([
          ('feature_extraction', EMGFeatureExtractor(fs=FS)), # Canonical extractor
          ('scaling', StandardScaler()),
          ('regression', clone(model_instance)) # Clone for safety
     ])
     with timer(f"Generating OOF predictions for {model_name}"):
          # Base arguments for cross_val_predict
          oof_cv_args = {
               'estimator': pipeline_oof_base, 'X': X_windowed, 'y': y_windowed,
               'cv': cv_strategy, 'n_jobs': N_JOBS, 'method': 'predict'
          }
          # Check metadata routing and add groups accordingly
          if get_config().get('enable_metadata_routing', False):
               oof_cv_args['params'] = {'groups': groups}
               print(f"  (Using metadata routing for groups in {model_name} OOF)")
          else:
               # Try passing groups directly if metadata routing is off
               # Older sklearn versions might raise TypeError if 'groups' is unexpected by predict
               try:
                   oof_cv_args['groups'] = groups
                   print(f"  (Passing groups directly for {model_name} OOF - Metadata routing disabled)")
                   oof_preds = cross_val_predict(**oof_cv_args)
               except TypeError as e:
                   if 'got an unexpected keyword argument \'groups\'' in str(e):
                        print(f"  Warning: Passing 'groups' directly failed for {model_name} OOF predict. Trying without explicit groups.")
                        # Remove groups argument and proceed
                        del oof_cv_args['groups']
                        oof_preds = cross_val_predict(**oof_cv_args)
                   else: raise e # Re-raise other TypeErrors

          # If the 'if' block executed or the first 'try' in 'else' succeeded
          if 'oof_preds' not in locals() or oof_preds is None:
               oof_preds = cross_val_predict(**oof_cv_args) # Run (either with params or corrected args)


          oof_predictions[model_name] = oof_preds
     print(f"  Shape of {model_name} OOF predictions: {oof_preds.shape}")
     oof_preds = None # Reset for safety in next loop iteration

# --- OOF Predictions for NeuralNetwork_v3 ---
print("\n--- Generating OOF Predictions for NeuralNetwork_v3 ---")
# Use the best configuration (best_cfg_nn_v3) and hyperparameters from Section 10.
# Input features are X_input_for_nn_cv (canonical features).

if 'best_cfg_nn_v3' not in locals() or best_cfg_nn_v3 is None:
    raise RuntimeError("Best config 'best_cfg_nn_v3' for NN-v3 not found. Run Section 10 first.")
if 'X_input_for_nn_cv' not in locals() or X_input_for_nn_cv is None:
     raise ValueError("NN features 'X_input_for_nn_cv' not found. Run Section 5.")

# Pre-allocate array/list for NN OOF predictions
nn_v3_oof_preds_array = np.zeros_like(y_windowed) # Shape (n_samples, n_targets)

fold_num_nn_v3_oof = 0
with timer("Generating OOF predictions for NeuralNetwork_v3 (Manual LOSO CV)"):
    # Iterate through CV splits using the canonical features
    for train_idx, val_idx in cv_strategy.split(X_input_for_nn_cv, y_windowed, groups=groups):
        fold_num_nn_v3_oof += 1
        print(f"  Processing NN-v3 OOF Fold {fold_num_nn_v3_oof}/{N_SPLITS}")

        # 1. Split feature data
        X_train_fold_nn_oof, X_val_fold_nn_oof = X_input_for_nn_cv[train_idx], X_input_for_nn_cv[val_idx]
        y_train_fold_nn_oof, y_val_fold_nn_oof = y_windowed[train_idx], y_windowed[val_idx]

        # 2. Scale features based on this fold's training data
        scaler_nn_oof_fold = StandardScaler()
        X_train_fold_nn_oof_scaled = scaler_nn_oof_fold.fit_transform(X_train_fold_nn_oof)
        X_val_fold_nn_oof_scaled = scaler_nn_oof_fold.transform(X_val_fold_nn_oof)

        # 3. Datasets & DataLoaders with explicit dtypes
        train_ds_nn_oof = TensorDataset(
            torch.tensor(X_train_fold_nn_oof_scaled, dtype=torch.float32),
            torch.tensor(y_train_fold_nn_oof, dtype=torch.float32)
        )
        val_ds_nn_oof = TensorDataset(
            torch.tensor(X_val_fold_nn_oof_scaled, dtype=torch.float32),
            torch.tensor(y_val_fold_nn_oof, dtype=torch.float32)
        )
        
        # Use the optimized DataLoader function
        train_loader_nn_oof = create_optimized_dataloader(train_ds_nn_oof, batch_size=BATCH_SIZE_NN_V3, shuffle=True)
        val_loader_nn_oof = create_optimized_dataloader(val_ds_nn_oof, batch_size=BATCH_SIZE_NN_V3 * 2, shuffle=False)

        # 4. Initialize model, loss, optimizer, scheduler, early stopping
        model_nn_v3_oof = EMG_NN(n_features=n_input_features_nn, n_joints=N_JOINTS,
                                 hidden_sizes=best_cfg_nn_v3['hidden'], dropout_p=best_cfg_nn_v3['dropout']).to(device)
        # Ensure the model is explicitly set to float32
        model_nn_v3_oof.float()
        
        criterion_nn_oof = nn.MSELoss()
        optimizer_nn_oof = torch.optim.Adam(model_nn_v3_oof.parameters(), lr=LR_NN_V3, weight_decay=WEIGHT_DECAY_NN_V3)
        scheduler_nn_oof = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer_nn_oof, mode='min', factor=0.5, patience=PATIENCE_LR_NN_V3, verbose=False)
        early_stopper_nn_oof = EarlyStop(patience=PATIENCE_ES_NN_V3)
        
        # Create GradScaler for mixed precision training
        scaler_nn_oof = GradScaler()

        # 5. Training loop for this fold
        print(f"    Training NN for OOF Fold {fold_num_nn_v3_oof}...")
        for epoch_oof in range(MAX_EPOCHS_NN_V3):
            # Use the previously defined train_nn_epoch function
            prefetcher = DataPrefetcher(train_loader_nn_oof, device)
            inputs, targets = prefetcher.next()
            batch_idx = 0
            
            model_nn_v3_oof.train()
            running_loss = 0.0
            
            while inputs is not None:
                batch_idx += 1
                
                # Zero the parameter gradients
                optimizer_nn_oof.zero_grad(set_to_none=True)
                
                # Mixed precision forward pass
                with autocast():
                    outputs = model_nn_v3_oof(inputs)
                    loss = criterion_nn_oof(outputs, targets)
                
                # Scaled backward pass
                scaler_nn_oof.scale(loss).backward()
                scaler_nn_oof.step(optimizer_nn_oof)
                scaler_nn_oof.update()
                
                running_loss += loss.item() * inputs.size(0)
                inputs, targets = prefetcher.next()
            
            train_loss = running_loss / len(train_loader_nn_oof.dataset)
            
            # Evaluation
            model_nn_v3_oof.eval()
            val_loss = 0.0
            all_preds = []
            all_targets = []
            
            prefetcher_val = DataPrefetcher(val_loader_nn_oof, device)
            inputs_val, targets_val = prefetcher_val.next()
            
            with torch.no_grad():
                while inputs_val is not None:
                    # Use autocast for consistency with training
                    with autocast():
                        outputs_val = model_nn_v3_oof(inputs_val)
                        loss_val = criterion_nn_oof(outputs_val, targets_val)
                    
                    val_loss += loss_val.item() * inputs_val.size(0)
                    all_preds.append(outputs_val.cpu().numpy())
                    all_targets.append(targets_val.cpu().numpy())
                    
                    inputs_val, targets_val = prefetcher_val.next()
            
            mse_val_oof = val_loss / len(val_loader_nn_oof.dataset)
            val_preds = np.concatenate(all_preds, axis=0)
            val_targets = np.concatenate(all_targets, axis=0)
            
            scheduler_nn_oof.step(mse_val_oof)
            if early_stopper_nn_oof(mse_val_oof):
                print(f"      OOF Fold {fold_num_nn_v3_oof} early stopping at epoch {epoch_oof+1}.")
                break

        # 6. Get PREDICTIONS on the validation set for this fold using the trained model
        model_nn_v3_oof.eval()
        val_preds_final = []
        
        prefetcher_final = DataPrefetcher(val_loader_nn_oof, device)
        inputs_final, _ = prefetcher_final.next()
        
        with torch.no_grad():
            while inputs_final is not None:
                with autocast():
                    outputs_final = model_nn_v3_oof(inputs_final)
                val_preds_final.append(outputs_final.cpu().numpy())
                inputs_final, _ = prefetcher_final.next()
                
        y_pred_fold_nn_v3_oof = np.concatenate(val_preds_final, axis=0)

        # 7. Store predictions in the correct rows of the pre-allocated array
        nn_v3_oof_preds_array[val_idx] = y_pred_fold_nn_v3_oof

        # Clean up GPU memory for the fold
        del model_nn_v3_oof, train_loader_nn_oof, val_loader_nn_oof, optimizer_nn_oof, scheduler_nn_oof, early_stopper_nn_oof
        if device == torch.device('cuda'): torch.cuda.empty_cache()
        gc.collect()

# Add the completed NN OOF predictions to the dictionary
oof_predictions["NeuralNetwork_v3"] = nn_v3_oof_preds_array
print(f"  Shape of NeuralNetwork_v3 OOF predictions: {oof_predictions['NeuralNetwork_v3'].shape}")

# Verify all intended base models have OOF predictions
ensemble_base_model_names = list(baseline_models_to_evaluate.keys()) + ["NeuralNetwork_v3"]
missing_oof = [name for name in ensemble_base_model_names if name not in oof_predictions or not isinstance(oof_predictions[name], np.ndarray)]
if missing_oof:
     raise RuntimeError(f"Missing or invalid OOF predictions for models: {missing_oof}")
else:
    print("\nAll required OOF predictions generated successfully.")

# %% [markdown]
# ### 11.2 Averaging Ensemble

# %%
print("\n--- Evaluating Averaging Ensemble ---")
# Identify the models whose OOF predictions will be averaged
model_names_for_avg = list(oof_predictions.keys()) # Should contain the 4 base models now
print(f"Averaging predictions from models: {model_names_for_avg}")

# Check shapes before summing
expected_shape = y_windowed.shape
for model_name in model_names_for_avg:
    if oof_predictions[model_name].shape != expected_shape:
        raise ValueError(f"Shape mismatch for OOF '{model_name}' ({oof_predictions[model_name].shape}) vs y_windowed ({expected_shape})")

# Calculate the average prediction
oof_preds_sum = np.sum([oof_predictions[name] for name in model_names_for_avg], axis=0)
oof_preds_avg = oof_preds_sum / len(model_names_for_avg)

# Evaluate the performance of the averaged OOF predictions
avg_rmse = root_mean_squared_error(y_windowed, oof_preds_avg)
avg_nmse = normalized_mean_squared_error(y_windowed, oof_preds_avg)
# Recalculate R2 score correctly using the OOF average predictions
avg_r2 = r2_score(y_windowed, oof_preds_avg)

print(f"\nAveraging Ensemble Performance (evaluated on OOF predictions):")
print(f"  RMSE: {avg_rmse:.4f}")
print(f"  NMSE: {avg_nmse:.4f}")
print(f"  R2 Score: {avg_r2:.4f}") # This should now be a meaningful R2 score

# Add results to summary DataFrame
# Remove old entry if exists to prevent duplication
results_summary_list = [item for item in results_summary_list if item['Model'] != 'AverageEnsemble']
results_summary_list.append({
    'Model': 'AverageEnsemble', 'Type': 'Ensemble',
    'RMSE_mean': avg_rmse, 'RMSE_std': np.nan, # Stddev not applicable to single OOF evaluation
    'NMSE_mean': avg_nmse, 'NMSE_std': np.nan,
    'R2_mean': avg_r2, 'R2_std': np.nan,
    'FitTime': np.nan # Fit time not applicable here
})
results_df = pd.DataFrame(results_summary_list) # Update results_df
# %% [markdown]
# ### 11.3 Stacking Ensemble (Meta-Learner)

# %%
print("\n--- Evaluating Stacking Ensemble (Multiple Meta-Learners) ---")

# 1. Create Meta-Features (X_meta) by concatenating OOF predictions
# Ensure the order of concatenation matches the 'model_names_for_stacking' list
model_names_for_stacking = list(oof_predictions.keys()) # Order matters here!
print(f"Stacking OOF predictions from models (in this order): {model_names_for_stacking}")

X_meta_list = []
for model_name in model_names_for_stacking:
    if model_name not in oof_predictions or oof_predictions[model_name] is None:
        raise ValueError(f"Missing OOF predictions for model: {model_name}")
    if oof_predictions[model_name].shape[0] != len(y_windowed):
         raise ValueError(f"OOF prediction shape mismatch for {model_name}. Expected {len(y_windowed)} samples, got {oof_predictions[model_name].shape[0]}.")
    X_meta_list.append(oof_predictions[model_name])

try:
    X_meta = np.hstack(X_meta_list)
    print(f"Successfully created Meta-Features (X_meta). Shape: {X_meta.shape}")
    # Expected shape: (n_samples, n_base_models * n_targets) -> (9185, 4 * 51) = (9185, 204)
    expected_cols = len(model_names_for_stacking) * N_JOINTS
    if X_meta.shape != (len(y_windowed), expected_cols):
        print(f"Warning: X_meta shape {X_meta.shape} differs from expected ({len(y_windowed)}, {expected_cols})")
except ValueError as e:
    print(f"Error creating X_meta by hstacking OOF predictions: {e}")
    for i, arr in enumerate(X_meta_list): print(f"  Shape of OOF for {model_names_for_stacking[i]}: {arr.shape}")
    raise
if 'X_meta' not in locals() or X_meta is None: raise RuntimeError("X_meta creation failed.")

# 2. Define Meta-Learners
# Use relatively simple meta-learners to avoid overfitting the OOF predictions
meta_learners = {
    "Stacking_Ridge": Ridge(alpha=1.0, random_state=RANDOM_STATE),
    "Stacking_Lasso": Lasso(alpha=0.01, max_iter=5000, tol=5e-4, random_state=RANDOM_STATE), # Increased max_iter/tol
    "Stacking_RF": RandomForestRegressor(n_estimators=50, max_depth=7, # Shallower depth for meta
                                         min_samples_split=10, min_samples_leaf=5,
                                         max_features=0.5, # Consider half of base model preds per split
                                         n_jobs=N_JOBS, random_state=RANDOM_STATE)
}
print(f"\nMeta-learners to evaluate: {list(meta_learners.keys())}")

# 3. Cross-Validate each Meta-Learner using LOSO CV on X_meta
# Remove previous specific stacking results if re-running cell
for meta_key in meta_learners.keys():
    results_summary_list = [item for item in results_summary_list if item['Model'] != meta_key]

for meta_name, meta_model in meta_learners.items():
    print(f"\n--- Cross-validating Meta-Learner: {meta_name} ---")
    # Define the pipeline for the meta-learner
    stacking_pipeline = Pipeline([
        # Optional: ('meta_scaling', StandardScaler()), # Consider if base model outputs vary significantly in scale
        ('meta_regression', clone(meta_model)) # Use the cloned meta-learner
    ])

    with timer(f"Cross-validating {meta_name}"):
        # Prepare base arguments for cross_validate
        base_cv_args = {
            'estimator': stacking_pipeline, 'X': X_meta, 'y': y_windowed,
            'cv': cv_strategy, 'scoring': scorers, 'n_jobs': N_JOBS,
            'return_train_score': False, 'error_score': 'raise'
        }

        stacking_scores = None # Initialize scores variable

        # Attempt cross-validation, handling potential 'groups' argument issues
        try:
            if get_config().get('enable_metadata_routing', False):
                # Try with metadata routing first
                print(f"  (Attempting CV with metadata routing for groups in {meta_name})")
                current_cv_args = {**base_cv_args, 'params': {'groups': groups}}
                stacking_scores = cross_validate(**current_cv_args)
                print("    CV with metadata routing successful.")
            else:
                # Try passing groups directly if metadata routing is off
                print(f"  (Attempting CV by passing groups directly for {meta_name})")
                current_cv_args = {**base_cv_args, 'groups': groups}
                stacking_scores = cross_validate(**current_cv_args)
                print("    CV with direct groups successful.")

        except TypeError as e:
            # Check if the error is specifically about the 'groups' argument
            if 'got an unexpected keyword argument \'groups\'' in str(e) or 'fit() got an unexpected keyword argument \'groups\'' in str(e):
                 print("  Metadata routing/direct groups failed. Falling back to manual CV loop (no groups passed to fit/predict).")
                 # Manual CV loop - This loop itself *implicitly* uses the groups via cv_strategy.split
                 fold_scores_stack = {key: [] for key in list(scorers.keys()) + ['fit_time', 'score_time']}
                 for fold_idx, (train_idx, test_idx) in enumerate(cv_strategy.split(X_meta, y_windowed, groups=groups)): # Groups used here for splitting
                     start_fit_time = time.time()
                     # Fit pipeline on fold data (NO groups argument passed here)
                     fitted_pipeline = clone(stacking_pipeline).fit(X_meta[train_idx], y_windowed[train_idx])
                     end_fit_time = time.time()
                     y_pred_meta = fitted_pipeline.predict(X_meta[test_idx])
                     end_score_time = time.time()

                     fold_scores_stack['fit_time'].append(end_fit_time - start_fit_time)
                     fold_scores_stack['score_time'].append(end_score_time - end_fit_time)

                     # Calculate scores manually for the fold
                     y_test_meta = y_windowed[test_idx] # Ground truth for the fold
                     for score_name, scorer_obj in scorers.items():
                          # Use scorer object directly if possible (newer sklearn), else fallback
                          try:
                              # Pass estimator, X_test, y_test
                              score_value = scorer_obj(fitted_pipeline, X_meta[test_idx], y_test_meta)
                          except TypeError: # If scorer expects y_true, y_pred
                               try:
                                   score_value = scorer_obj._score_func(y_test_meta, y_pred_meta, **scorer_obj._kwargs)
                               except Exception as score_err:
                                   print(f"Error calculating score {score_name} manually: {score_err}")
                                   score_value = np.nan # Assign NaN if scoring fails

                          # Manual loop calculates the direct metric value (e.g., positive RMSE)
                          # We need to store it in a way compatible with cross_validate's output later
                          fold_scores_stack[score_name].append(score_value)


                 # Reconstruct scores dictionary like cross_validate's output
                 stacking_scores = {}
                 stacking_scores['fit_time'] = np.array(fold_scores_stack['fit_time'])
                 stacking_scores['score_time'] = np.array(fold_scores_stack['score_time'])
                 for score_name, scorer_obj in scorers.items():
                     # Apply the sign convention used by cross_validate based on greater_is_better
                     sign = getattr(scorer_obj, '_sign', 1) # Default sign is 1 (maximize)
                     raw_scores = np.array(fold_scores_stack[score_name])
                     # Ensure scores are stored with the correct sign (e.g., -RMSE, -NMSE, +R2)
                     stacking_scores[f"test_{score_name}"] = raw_scores * sign
                 print("    Manual CV loop finished.")

            else:
                 # If the TypeError was not related to 'groups', re-raise it
                 print(f"  An unexpected TypeError occurred during cross-validation: {e}")
                 raise e

    # --- End of CV attempt for this meta-learner ---

    # Check if stacking_scores were successfully obtained
    if stacking_scores is None:
        raise RuntimeError(f"Cross-validation failed to produce scores for {meta_name}.")

    # Store and display results for this meta-learner
    all_cv_results[meta_name] = stacking_scores # Store detailed scores

    # Calculate mean/std, remembering that RMSE/NMSE are stored negatively
    mean_rmse = -np.nanmean(stacking_scores['test_rmse']); std_rmse = np.nanstd(stacking_scores['test_rmse'])
    mean_nmse = -np.nanmean(stacking_scores['test_nmse']); std_nmse = np.nanstd(stacking_scores['test_nmse'])
    mean_r2 = np.nanmean(stacking_scores['test_r2']); std_r2 = np.nanstd(stacking_scores['test_r2'])
    fit_time_mean = np.nanmean(stacking_scores['fit_time'])

    print(f"\n{meta_name} Performance (Cross-Validated on OOF):")
    print(f"  RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}")
    print(f"  NMSE: {mean_nmse:.4f} +/- {std_nmse:.4f}")
    print(f"  R2 Score: {mean_r2:.4f} +/- {std_r2:.4f}")
    print(f"  Avg Meta Fit Time: {fit_time_mean:.2f}s")

    # Append summary results
    results_summary_list.append({
        'Model': meta_name, 'Type': 'Ensemble',
        'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse,
        'NMSE_mean': mean_nmse, 'NMSE_std': std_nmse,
        'R2_mean': mean_r2, 'R2_std': std_r2,
        'FitTime': fit_time_mean
    })

# Update the main results DataFrame after evaluating all meta-learners
results_df = pd.DataFrame(results_summary_list)

print("\n--- Updated Summary Table (including Stacking) ---")
print(results_df.sort_values(by='RMSE_mean').to_string(index=False, float_format="%.4f"))

# %% [markdown]
# ### 11.4 Meta-Learner Contribution Analysis (Stacking)

# %%
print("\n--- Stacking Meta-Learner Contribution Analysis (using Ridge) ---")

# Choose a linear meta-learner (e.g., Stacking_Ridge)
analysis_meta_learner = clone(meta_learners["Stacking_Ridge"]) # Analyze Ridge coefficients

# Train the chosen meta-learner on the full OOF dataset (X_meta)
print(f"Training {analysis_meta_learner.__class__.__name__} on full X_meta for coefficient analysis...")
with timer("Fitting meta-learner on full X_meta"):
    # Ensure X_meta exists from the previous cell
    if 'X_meta' not in locals() or X_meta is None:
        raise NameError("X_meta not found. Run the previous cell (11.3) first.")
    analysis_meta_learner.fit(X_meta, y_windowed)

# Check if the model has coefficients (linear models do)
if hasattr(analysis_meta_learner, 'coef_'):
    coefficients = analysis_meta_learner.coef_
    # Coef shape is (n_targets, n_meta_features) = (51, num_base_models * 51)
    print(f"Shape of meta-learner coefficients: {coefficients.shape}")

    # Calculate average absolute coefficient magnitude for each base model's block of features
    n_base_models = len(model_names_for_stacking)
    n_joints_meta = N_JOINTS # Should be 51
    avg_coef_magnitudes = {}

    print("\nAverage Absolute Coefficient Magnitude per Base Model:")
    for i, model_name in enumerate(model_names_for_stacking):
        start_col = i * n_joints_meta
        end_col = (i + 1) * n_joints_meta
        # Extract the block of coefficients corresponding to this base model
        model_coefs = coefficients[:, start_col:end_col]
        # Calculate the average absolute magnitude across all targets and features for this model
        avg_mag = np.mean(np.abs(model_coefs))
        avg_coef_magnitudes[model_name] = avg_mag
        print(f"  {model_name}: {avg_mag:.4f}")

    # Create a Series for easy sorting/viewing
    coef_mag_series = pd.Series(avg_coef_magnitudes).sort_values(ascending=False)
    print("\nRelative Contribution based on Avg. Abs. Coefficient Magnitude:")
    print(coef_mag_series)

    # Plotting the contributions
    plt.figure(figsize=(8, 5))
    coef_mag_series.plot(kind='bar')
    plt.title(f'Stacking ({analysis_meta_learner.__class__.__name__}): Base Model Contribution')
    plt.ylabel('Average Absolute Coefficient Magnitude')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

else:
    print(f"Selected meta-learner {analysis_meta_learner.__class__.__name__} does not have 'coef_' attribute. Skipping contribution analysis.")
# %% [markdown]
# ## 12. Objective 7: Final Model Selection and Test Set Prediction Preparation

# %%
# Add optimized DataLoader creation function
def create_optimized_dataloader(dataset, batch_size, shuffle=True, num_workers=NUM_WORKERS_LOADER, 
                               pin_memory=True if torch.cuda.is_available() else False):
    return DataLoader(
        dataset, 
        batch_size=batch_size, 
        shuffle=shuffle, 
        num_workers=num_workers, 
        pin_memory=pin_memory,
        persistent_workers=True if num_workers > 0 else False,
        prefetch_factor=2 if num_workers > 0 else None
    )

# Keep original DataPrefetcher class for optimized data transfers
class DataPrefetcher:
    def __init__(self, loader, device):
        self.loader = iter(loader)
        self.device = device
        self.stream = torch.cuda.Stream()
        self.preload()
        
    def preload(self):
        try:
            self.next_input, self.next_target = next(self.loader)
        except StopIteration:
            self.next_input = None
            self.next_target = None
            return
        
        with torch.cuda.stream(self.stream):
            if isinstance(self.next_input, tuple):
                self.next_input = self.next_input[0].to(self.device, non_blocking=True).float()
            else:
                self.next_input = self.next_input.to(self.device, non_blocking=True).float()
                
            if self.next_target is not None:
                self.next_target = self.next_target.to(self.device, non_blocking=True).float()
            
    def next(self):
        torch.cuda.current_stream().wait_stream(self.stream)
        input = self.next_input
        target = self.next_target
        self.preload()
        return input, target

print("\n--- Objective 7: Final Model Selection and Preparation ---")

# Ensure results_df contains all evaluation results up to this point
if 'results_df' not in globals() or results_df.empty:
    raise RuntimeError("Results DataFrame `results_df` is not defined or empty. Run previous evaluation cells.")

print("\nFinal Combined Model Comparison Table (sorted by RMSE):")
# Display the full comparison table, sorted by the primary metric (RMSE)
print(results_df.sort_values(by='RMSE_mean').to_string(index=False, float_format="%.4f"))

# 1. Select the best overall model based on Mean CV RMSE
try:
    best_model_row = results_df.loc[results_df['RMSE_mean'].idxmin()]
    best_overall_model_name = best_model_row['Model']
    best_overall_rmse = best_model_row['RMSE_mean']
    print(f"\nSelected Best Overall Model for Guided Gestures Prediction: '{best_overall_model_name}' (Mean CV RMSE: {best_overall_rmse:.4f})")
except KeyError:
    raise RuntimeError("Could not determine best model. 'RMSE_mean' or 'Model' key might be missing in results_df.")
except Exception as e:
    raise RuntimeError(f"Error selecting best model from results_df: {e}")

# 2. Prepare the final model object for prediction
# This object (`final_model_for_prediction_guided`) will be structured depending on the chosen model type.
final_model_for_prediction_guided = None

# Case 1: Neural Network (NN_v3) was chosen
if best_overall_model_name == 'NeuralNetwork_v3':
    print(f"\nPreparing '{best_overall_model_name}' (already trained on full data in Section 10).")
    # The final model and scaler were trained and saved in Section 10
    # We just need to load/reference them. Use the dictionary created at the end of Sec 10.
    if 'final_trained_model_dict_nn_v3' in locals():
        final_model_for_prediction_guided = final_trained_model_dict_nn_v3
        # Verify the components exist
        if not all(k in final_model_for_prediction_guided for k in ['model_type', 'model', 'scaler', 'feature_extractor']):
             raise ValueError("final_trained_model_dict_nn_v3 is missing required keys.")
        print(f"'{best_overall_model_name}' is ready for prediction.")
    else:
        raise NameError("The dictionary 'final_trained_model_dict_nn_v3' containing the final NN model/scaler was not found. Ensure Section 10 completed successfully.")

# Case 2: A Stacking Ensemble was chosen
elif best_overall_model_name in meta_learners: # Check if the name matches one of the stacking keys
    print(f"\nPreparing final Stacking Ensemble: '{best_overall_model_name}'...")
    # Requires:
    #   a) All base models trained on 100% of the training data.
    #   b) The chosen meta-learner trained on the original OOF predictions (X_meta).

    # a) Train/Load Base Models on Full Data
    trained_final_base_models_for_stacking = {}
    print("  Retraining/preparing base models on full data...")
    # Sklearn base models (Ridge, Lasso, RF)
    for sk_model_name, sk_model_instance in baseline_models_to_evaluate.items():
        p_base = Pipeline([('feature_extraction', EMGFeatureExtractor(fs=FS)), # Canonical extractor
                           ('scaling', StandardScaler()),
                           ('regression', clone(sk_model_instance))])
        with timer(f"    Fitting final base model '{sk_model_name}' for stacking"):
            p_base.fit(X_windowed, y_windowed) # Train on all raw windowed data
        trained_final_base_models_for_stacking[sk_model_name] = p_base

    # NeuralNetwork_v3 base model (use the one already trained on full data)
    if 'final_trained_model_dict_nn_v3' in locals():
        trained_final_base_models_for_stacking['NeuralNetwork_v3'] = final_trained_model_dict_nn_v3 # Use the dict containing model, scaler, etc.
        print("    Using pre-trained NN-v3 (from Sec 10) as a base model.")
    else:
        raise NameError("Final NN-v3 model needed for stacking base not found. Run Section 10.")

    # b) Train Meta-Learner on X_meta
    final_meta_learner_instance = clone(meta_learners[best_overall_model_name])
    # Meta pipeline usually just contains the regressor
    final_stacking_meta_pipeline = Pipeline([('meta_regression', final_meta_learner_instance)])

    if 'X_meta' not in locals() or X_meta is None:
        raise NameError("X_meta (OOF predictions for training meta-learner) not found. Run Section 11.3.")
    if 'model_names_for_stacking' not in locals() or not model_names_for_stacking:
        raise NameError("'model_names_for_stacking' (order of OOF) not found. Run Section 11.3.")

    print(f"  Retraining meta-learner '{best_overall_model_name}' on full X_meta...")
    with timer(f"    Fitting final meta-learner '{best_overall_model_name}'"):
        final_stacking_meta_pipeline.fit(X_meta, y_windowed)

    # Store all components needed for prediction
    final_model_for_prediction_guided = {
        'model_type': 'StackingEnsemble',
        'name': best_overall_model_name,
        'meta_model_pipeline': final_stacking_meta_pipeline, # The trained meta-learner pipeline
        'base_models_trained': trained_final_base_models_for_stacking, # Dict of fully trained base models/pipelines
        'base_model_order': model_names_for_stacking # Crucial: Order for creating meta-features at prediction time
    }
    print(f"Final Stacking Ensemble '{best_overall_model_name}' prepared.")

# Case 3: Averaging Ensemble was chosen
elif best_overall_model_name == 'AverageEnsemble':
    print(f"\nPreparing final Average Ensemble...")
    # Requires all base models trained on 100% of the training data.

    trained_final_base_models_for_avg = {}
    print("  Retraining/preparing base models on full data...")
    # Sklearn base models
    for sk_model_name, sk_model_instance in baseline_models_to_evaluate.items():
        p_avg = Pipeline([('feature_extraction', EMGFeatureExtractor(fs=FS)),
                          ('scaling', StandardScaler()),
                          ('regression', clone(sk_model_instance))])
        with timer(f"    Fitting final base model '{sk_model_name}' for averaging"):
            p_avg.fit(X_windowed, y_windowed)
        trained_final_base_models_for_avg[sk_model_name] = p_avg

    # NN-v3 base model
    if 'final_trained_model_dict_nn_v3' in locals():
        trained_final_base_models_for_avg['NeuralNetwork_v3'] = final_trained_model_dict_nn_v3
        print("    Using pre-trained NN-v3 (from Sec 10) as a base model.")
    else:
        raise NameError("Final NN-v3 model needed for averaging base not found. Run Section 10.")

    if 'model_names_for_avg' not in locals() or not model_names_for_avg:
        raise NameError("'model_names_for_avg' (from Sec 11.2) not found.")

    # Store components needed for prediction
    final_model_for_prediction_guided = {
        'model_type': 'AverageEnsemble',
        'base_models_trained': trained_final_base_models_for_avg, # Dict of fully trained base models/pipelines
        'base_model_order': model_names_for_avg # List of models to average
    }
    print("Final Average Ensemble prepared.")

# Case 4: A Baseline or SFS Pipeline was chosen
elif best_overall_model_name in baseline_models_to_evaluate or best_overall_model_name.endswith("_SFS"):
    print(f"\nRetraining final pipeline: '{best_overall_model_name}' on all Guided Training Data...")

    if best_overall_model_name.endswith("_SFS"):
        # Reconstruct the SFS pipeline
        base_model_name_sfs = best_overall_model_name.replace("_SFS", "")
        if base_model_name_sfs not in baseline_models_to_evaluate:
            raise ValueError(f"Base model '{base_model_name_sfs}' for SFS pipeline not found.")
        final_regressor_instance = clone(baseline_models_to_evaluate[base_model_name_sfs])

        if 'selected_feature_types_only' not in locals() or not selected_feature_types_only:
            raise NameError("'selected_feature_types_only' from SFS (Section 9) not found or empty.")

        final_model_for_prediction_guided = Pipeline([
            ('feature_extraction', EMGFeatureExtractor(fs=FS, selected_features=selected_feature_types_only)),
            ('scaling', StandardScaler()),
            ('regression', final_regressor_instance)
        ])
        feature_info = f"({len(selected_feature_types_only)} selected feature types)"
    else:
        # Reconstruct the baseline pipeline (all features)
        final_regressor_instance = clone(baseline_models_to_evaluate[best_overall_model_name])
        final_model_for_prediction_guided = Pipeline([
            ('feature_extraction', EMGFeatureExtractor(fs=FS)), # Canonical extractor
            ('scaling', StandardScaler()),
            ('regression', final_regressor_instance)
        ])
        feature_info = "(all features)"

    with timer(f"Fitting final pipeline '{best_overall_model_name}' {feature_info}"):
        # Train the pipeline on the full raw windowed EMG data
        final_model_for_prediction_guided.fit(X_windowed, y_windowed)

    print(f"Final pipeline '{best_overall_model_name}' retrained.")

# Fallback: Should not happen if results_df is correct
else:
    raise ValueError(f"Don't know how to prepare the final model for '{best_overall_model_name}'. Check model name and add logic if necessary.")

# Final check
if final_model_for_prediction_guided is None:
    raise RuntimeError(f"Failed to prepare the final model '{best_overall_model_name}' for prediction. Check logic.")

print(f"\n'{best_overall_model_name}' is fully prepared/retrained and stored in 'final_model_for_prediction_guided'. Ready for test set prediction.")

# --- Saving the final chosen model object ---
# This saves the actual object (pipeline or dictionary) needed for prediction.
SAVE_FINAL_MODEL_OBJECT = True
if SAVE_FINAL_MODEL_OBJECT:
    # Create a filename based on the chosen model
    safe_model_name = best_overall_model_name.lower().replace(' ', '_').replace('/', '_').replace('.', '')
    save_filename_guided = f"final_guided_model_object_{safe_model_name}.pkl"

    # NN model was already saved as .pth/.pkl, the 'dict' is what we might save here if needed,
    # but typically we load from .pth/.pkl for NN prediction. Let's skip saving the dict itself.
    if isinstance(final_model_for_prediction_guided, dict) and final_model_for_prediction_guided.get('model_type') == 'NeuralNetwork_v3':
         print(f"Skipping saving NN dictionary object. Use saved '{MODEL_NN_V3_PATH}' and '{SCALER_NN_V3_PATH}' for prediction.")
    else: # Save sklearn pipelines or ensemble dictionaries
        try:
            with open(save_filename_guided, 'wb') as f:
                pickle.dump(final_model_for_prediction_guided, f)
            print(f"✅ Final chosen model object ('{best_overall_model_name}') saved to: {save_filename_guided}")
        except Exception as e:
            print(f"⚠️ Error saving final model object '{best_overall_model_name}': {e}")
            # Print details for debugging if save fails
            print(f"   Object type: {type(final_model_for_prediction_guided)}")
            if isinstance(final_model_for_prediction_guided, dict):
                print(f"   Object keys: {final_model_for_prediction_guided.keys()}")

# %% [markdown]
# ## 13. Generate Test Set Predictions (Guided Gestures)

# %%
# Import mixed precision for inference
from torch.cuda.amp import autocast

# Updated DataPrefetcher class that handles lists, tensors, and multiple cases
class TestDataPrefetcher:
    def __init__(self, loader, device):
        self.loader = iter(loader)
        self.device = device
        self.stream = torch.cuda.Stream()
        self.preload()
        
    def preload(self):
        try:
            next_data = next(self.loader)
            if isinstance(next_data, tuple) and len(next_data) == 2:
                self.next_input, self.next_target = next_data
            else:
                self.next_input = next_data
                self.next_target = None
        except StopIteration:
            self.next_input = None
            self.next_target = None
            return
        
        with torch.cuda.stream(self.stream):
            # Handle different input types
            if self.next_input is not None:
                if isinstance(self.next_input, list):
                    # Convert list to tensor if needed
                    self.next_input = torch.tensor(self.next_input, device=self.device, dtype=torch.float32)
                elif isinstance(self.next_input, tuple):
                    self.next_input = self.next_input[0].to(self.device, non_blocking=True).float()
                elif isinstance(self.next_input, torch.Tensor):
                    self.next_input = self.next_input.to(self.device, non_blocking=True).float()
                else:
                    print(f"Warning: Unexpected input type: {type(self.next_input)}")
                    
            # Handle target if present
            if self.next_target is not None:
                if isinstance(self.next_target, list):
                    self.next_target = torch.tensor(self.next_target, device=self.device, dtype=torch.float32)
                elif isinstance(self.next_target, torch.Tensor):
                    self.next_target = self.next_target.to(self.device, non_blocking=True).float()
            
    def next(self):
        torch.cuda.current_stream().wait_stream(self.stream)
        input_data = self.next_input
        target_data = self.next_target
        self.preload()
        
        if target_data is not None:
            return input_data, target_data
        else:
            return input_data

print("\n--- Generating Predictions for Guided Test Set ---")

# Set CUDA optimization flags
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

# 1. Load the test data
X_TEST_GUIDED_PATH = "guided_testset_X.npy"
with timer("Loading Guided Test Data"):
    X_test_guided_raw = np.load(X_TEST_GUIDED_PATH) # Shape: (n_sessions, n_windows_per_session, n_electrodes, window_size)
    print(f"Raw Guided Test Data Shape: {X_test_guided_raw.shape}")

# Reshape test data to match the expected input format: (total_windows, n_electrodes, window_size)
n_test_sessions, n_test_windows_per_session, n_test_electrodes, test_window_size = X_test_guided_raw.shape
X_test_guided_reshaped = X_test_guided_raw.reshape(-1, n_test_electrodes, test_window_size)
print(f"Reshaped Guided Test Data Shape: {X_test_guided_reshaped.shape}")
# Expected total windows: 5 sessions * 332 windows/session = 1660
n_total_test_windows_guided = n_test_sessions * n_test_windows_per_session
if X_test_guided_reshaped.shape[0] != n_total_test_windows_guided:
     print(f"Warning: Reshaped test window count ({X_test_guided_reshaped.shape[0]}) doesn't match expected ({n_total_test_windows_guided}).")

# Check consistency
if n_test_electrodes != N_ELECTRODES or test_window_size != WINDOW_SIZE:
    raise ValueError("Test data dimensions (electrodes, window_size) mismatch training data.")

# 2. Make predictions using the chosen final model
guided_predictions = None

with timer(f"Predicting with final model '{best_overall_model_name}'"):

    # Case A: Sklearn Pipeline (Baseline, SFS)
    if isinstance(final_model_for_prediction_guided, Pipeline):
        print("  Predicting using scikit-learn Pipeline...")
        # The pipeline handles feature extraction and scaling internally
        guided_predictions = final_model_for_prediction_guided.predict(X_test_guided_reshaped)

    # Case B: Neural Network (NN_v3)
    elif isinstance(final_model_for_prediction_guided, dict) and final_model_for_prediction_guided.get('model_type') == 'NeuralNetwork_v3':
        print("  Predicting using NeuralNetwork_v3...")
        nn_model = final_model_for_prediction_guided['model']
        nn_scaler = final_model_for_prediction_guided['scaler']
        nn_extractor = final_model_for_prediction_guided['feature_extractor']

        # 1. Extract features from raw test EMG windows
        print("    Extracting features from test data...")
        X_test_guided_features = nn_extractor.fit_transform(X_test_guided_reshaped) # Fit shouldn't be strictly necessary if fitted before, but safe

        # 2. Scale features using the scaler fitted on training data
        print("    Scaling test features...")
        X_test_guided_features_scaled = nn_scaler.transform(X_test_guided_features)

        # 3. Predict using the NN model with optimized inference
        print("    Running NN prediction...")
        nn_model.eval() # Set to evaluation mode
        
        # Create optimized dataset with explicit dtype
        test_dataset_nn = TensorDataset(
            torch.tensor(X_test_guided_features_scaled, dtype=torch.float32)
        )
        
        # Use larger batch size and optimized dataloader for inference
        test_loader_nn = create_optimized_dataloader(
            test_dataset_nn, 
            batch_size=BATCH_SIZE_NN_V3 * 4,  # Larger batches for inference
            shuffle=False, 
            num_workers=NUM_WORKERS_LOADER
        )

        # Use standard prediction approach instead of prefetcher
        preds_list_nn = []
        with torch.no_grad():
            for batch in test_loader_nn:
                inputs = batch[0].to(device, non_blocking=True).float()
                with autocast():
                    outputs = nn_model(inputs)
                preds_list_nn.append(outputs.cpu().numpy())

        guided_predictions = np.concatenate(preds_list_nn, axis=0)

    # Case C: Averaging Ensemble
    elif isinstance(final_model_for_prediction_guided, dict) and final_model_for_prediction_guided.get('model_type') == 'AverageEnsemble':
        print("  Predicting using AverageEnsemble...")
        base_models = final_model_for_prediction_guided['base_models_trained']
        base_order = final_model_for_prediction_guided['base_model_order']
        test_preds_sum = np.zeros((X_test_guided_reshaped.shape[0], N_JOINTS))

        for model_name in base_order:
            print(f"    Getting predictions from base model: {model_name}")
            base_model_obj = base_models[model_name]
            if isinstance(base_model_obj, Pipeline): # Sklearn pipeline base model
                test_preds_base = base_model_obj.predict(X_test_guided_reshaped)
            elif isinstance(base_model_obj, dict) and base_model_obj.get('model_type') == 'NeuralNetwork_v3': # NN base model
                # Predict using the NN base model logic (extract, scale, predict)
                nn_model_base = base_model_obj['model']
                nn_scaler_base = base_model_obj['scaler']
                nn_extractor_base = base_model_obj['feature_extractor']
                X_test_feat_base = nn_extractor_base.fit_transform(X_test_guided_reshaped)
                X_test_feat_scaled_base = nn_scaler_base.transform(X_test_feat_base)
                
                nn_model_base.eval()
                
                # Create optimized dataset with explicit dtype
                ds_base = TensorDataset(
                    torch.tensor(X_test_feat_scaled_base, dtype=torch.float32)
                )
                
                # Use optimized dataloader for inference
                loader_base = create_optimized_dataloader(
                    ds_base, 
                    batch_size=BATCH_SIZE_NN_V3 * 4,  
                    shuffle=False
                )
                
                # Use standard approach instead of prefetcher
                preds_list_base = []
                with torch.no_grad():
                    for batch in loader_base:
                        inputs = batch[0].to(device, non_blocking=True).float()
                        with autocast():
                            outputs_b = nn_model_base(inputs)
                        preds_list_base.append(outputs_b.cpu().numpy())
                        
                test_preds_base = np.concatenate(preds_list_base, axis=0)
            else:
                raise TypeError(f"Unsupported base model type for averaging: {type(base_model_obj)}")
            test_preds_sum += test_preds_base
        guided_predictions = test_preds_sum / len(base_order)

    # Case D: Stacking Ensemble
    elif isinstance(final_model_for_prediction_guided, dict) and final_model_for_prediction_guided.get('model_type') == 'StackingEnsemble':
        print(f"  Predicting using StackingEnsemble '{final_model_for_prediction_guided['name']}'...")
        meta_pipeline = final_model_for_prediction_guided['meta_model_pipeline']
        base_models = final_model_for_prediction_guided['base_models_trained']
        base_order = final_model_for_prediction_guided['base_model_order']

        # 1. Get predictions from each fully trained base model on the test set
        print("    Getting base model predictions on test data...")
        test_meta_features_list = []
        for model_name in base_order:
            print(f"      Base model: {model_name}")
            base_model_obj = base_models[model_name]
            if isinstance(base_model_obj, Pipeline): # Sklearn pipeline base model
                test_preds_base = base_model_obj.predict(X_test_guided_reshaped)
            elif isinstance(base_model_obj, dict) and base_model_obj.get('model_type') == 'NeuralNetwork_v3': # NN base model
                # Predict using the NN base model logic
                nn_model_base = base_model_obj['model']
                nn_scaler_base = base_model_obj['scaler']
                nn_extractor_base = base_model_obj['feature_extractor']
                X_test_feat_base = nn_extractor_base.fit_transform(X_test_guided_reshaped)
                X_test_feat_scaled_base = nn_scaler_base.transform(X_test_feat_base)
                
                nn_model_base.eval()
                
                # Create optimized dataset with explicit dtype
                ds_base = TensorDataset(
                    torch.tensor(X_test_feat_scaled_base, dtype=torch.float32)
                )
                
                # Use optimized dataloader
                loader_base = create_optimized_dataloader(
                    ds_base, 
                    batch_size=BATCH_SIZE_NN_V3 * 4, 
                    shuffle=False
                )
                
                # Use standard approach instead of prefetcher
                preds_list_base = []
                with torch.no_grad():
                    for batch in loader_base:
                        inputs = batch[0].to(device, non_blocking=True).float()
                        with autocast():
                            outputs_b = nn_model_base(inputs)
                        preds_list_base.append(outputs_b.cpu().numpy())
                
                test_preds_base = np.concatenate(preds_list_base, axis=0)
            else:
                raise TypeError(f"Unsupported base model type for stacking: {type(base_model_obj)}")
            test_meta_features_list.append(test_preds_base)

        # 2. Create meta-features for the test set
        X_meta_test = np.hstack(test_meta_features_list)
        print(f"    Created test meta-features. Shape: {X_meta_test.shape}")

        # 3. Predict using the trained meta-learner pipeline
        print("    Predicting with meta-learner...")
        guided_predictions = meta_pipeline.predict(X_meta_test)

    else:
        raise TypeError(f"Unsupported final model type for prediction: {type(final_model_for_prediction_guided)}")

# 3. Final Check of Prediction Shape
if guided_predictions is None:
    raise RuntimeError("Prediction failed, `guided_predictions` is None.")

print(f"\nShape of final guided predictions: {guided_predictions.shape}")
# Expected shape: (n_total_test_windows, n_joints) -> (1660, 51)
expected_guided_pred_shape = (n_total_test_windows_guided, N_JOINTS)
if guided_predictions.shape != expected_guided_pred_shape:
    print(f"WARNING: Final guided prediction shape {guided_predictions.shape} does not match expected {expected_guided_pred_shape}.")
else:
    print("Final guided prediction shape matches expected shape.")

# Note: The project requires concatenating guided and freemoves predictions later.
# This `guided_predictions` array is the first part needed for the final submission file.

# Clean up large test data array if needed
del X_test_guided_raw, X_test_guided_reshaped
if 'X_test_guided_features' in locals(): del X_test_guided_features
if 'X_test_guided_features_scaled' in locals(): del X_test_guided_features_scaled
if 'X_meta_test' in locals(): del X_meta_test
if torch.cuda.is_available(): torch.cuda.empty_cache()
gc.collect()

print("\nGuided Gestures prediction generation complete.")

# %% [markdown]
# ---
# # END OF GUIDED GESTURES PART
#
# The `guided_predictions` numpy array now holds the predictions for the guided gestures test set.
# The next steps would involve:
# 1. Repeating a similar process (potentially selecting a different best model) for the **freemoves dataset**.
# 2. Loading the freemoves test set and generating `freemoves_predictions`.
# 3. Concatenating `guided_predictions` and `freemoves_predictions` (`np.vstack`).
# 4. Saving the combined array to a CSV file for submission.
# ---
```

```python
# %% [markdown]
# # INFO-F-422 Project: EMG-based Hand Pose Estimation (Guided Gestures) - ENHANCED
#
# Team: [Your Team Name/Number]
# Members: [Your Names]
#
# This notebook implements and enhances approaches for predicting hand joint angles from sEMG signals.
# It includes:
# - Robust data loading, filtering, and windowing.
# - Comprehensive feature extraction.
# - Leave-One-Session-Out Cross-Validation.
# - Baseline models: Ridge, Lasso, RF.
# - Advanced models: XGBoost, LightGBM.
# - Sophisticated 3-layer Neural Network (NN_v3) with Optuna Hyperparameter Optimization.
# - Ensembling: Averaging and Stacking (with tuned meta-learners).
# - Detailed metric reporting and diagnostics.

# %%
# Core Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time
import pickle
from contextlib import contextmanager
import gc
import warnings

# Scikit-learn Modules
from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, RobustScaler # Added RobustScaler
from sklearn.model_selection import GroupKFold, cross_validate, train_test_split, cross_val_predict
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.metrics import mean_squared_error, make_scorer, r2_score
from sklearn.linear_model import Ridge, Lasso, RidgeCV, LassoCV # Added CV versions
from sklearn.ensemble import RandomForestRegressor
from sklearn import set_config, get_config

# Gradient Boosting Machines
import xgboost as xgb
import lightgbm as lgb

# PyTorch Modules
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
import torch.nn.functional as F
from torch.cuda.amp import autocast, GradScaler # For mixed-precision

# Optuna for Hyperparameter Optimization
import optuna

# SciPy
import scipy.fftpack
from scipy.signal import butter, filtfilt

# Configure visualizations and settings
%matplotlib inline
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
warnings.filterwarnings("ignore", category=UserWarning, module="torch.optim.lr_scheduler")
warnings.filterwarnings("ignore", category=FutureWarning) # Suppress some numpy/sklearn future warnings

# --- Global Constants and Configuration ---
RANDOM_STATE = 42
N_JOBS = -1 # Use all available CPU cores for sklearn jobs (-1) or set to a specific number
FS = 1024   # Sampling frequency from documentation

# Optuna HPO Settings (adjust for quicker/longer runs)
N_OPTUNA_TRIALS_NN = 25 # Number of HPO trials for Neural Network (e.g., 20-100)
N_OPTUNA_TRIALS_GBM = 15 # Number of HPO trials for GBMs (e.g., 10-50)
N_FOLDS_HPO_CV = 3     # Number of folds for CV within HPO objective (faster than full 5-fold)

# NN Training Settings
MAX_EPOCHS_NN_V3   = 100 # Reduced from 150 for potentially faster HPO/CV, can be increased
# BATCH_SIZE_NN_V3 will be tuned by Optuna
# LR_NN_V3 will be tuned by Optuna
PATIENCE_LR_SCHEDULER  = 8
PATIENCE_EARLY_STOPPING = 15
NUM_WORKERS_LOADER = 4 # Adjust based on your CPU cores and system

# Define a timing context manager
@contextmanager
def timer(label: str):
    start = time.time()
    print(f"{label}...")
    yield
    end = time.time()
    print(f"{label} done in {end - start:.2f} seconds.\n")

# Scikit-learn metadata routing
try:
    set_config(enable_metadata_routing=True)
    print(f"Scikit-learn metadata routing enabled: {get_config().get('enable_metadata_routing', False)}")
except Exception as e:
    print(f"Could not enable metadata routing (sklearn version might be older): {e}")

# PyTorch Device and CUDA Optimizations
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using PyTorch device: {device}")
if device == torch.device("cuda"):
    print(f"CUDA Device Name: {torch.cuda.get_device_name(0)}")
    torch.backends.cudnn.benchmark = True # Good if input sizes don't change much
    # For Ampere GPUs and newer (like RTX 30xx, 40xx, 50xx)
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True
    # torch.set_float32_matmul_precision('high') # PyTorch 1.12+ for TF32

# %% [markdown]
# ## 1. Load Data (Identical to Foucauld's)

# %%
X_TRAIN_VAL_PATH = "guided_dataset_X.npy"
Y_TRAIN_VAL_PATH = "guided_dataset_y.npy"

with timer("Loading Data"):
    X_train_val_raw = np.load(X_TRAIN_VAL_PATH)
    y_train_val_raw = np.load(Y_TRAIN_VAL_PATH)

N_SESSIONS_TRAIN_VAL = X_train_val_raw.shape[0]
N_ELECTRODES = X_train_val_raw.shape[1]
N_TIME_POINTS_TRAIN_VAL = X_train_val_raw.shape[2]
N_JOINTS = y_train_val_raw.shape[1]

print(f"X_train_val_raw: {X_train_val_raw.shape}, y_train_val_raw: {y_train_val_raw.shape}")
print(f"Sessions: {N_SESSIONS_TRAIN_VAL}, Electrodes: {N_ELECTRODES}, Time Points: {N_TIME_POINTS_TRAIN_VAL}, Joints: {N_JOINTS}")

# %% [markdown]
# ## 2. Signal Filtering (Identical to Foucauld's)

# %%
def butter_bandpass(lowcut, highcut, fs, order=5):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    if low <= 0 or high >= 1:
        raise ValueError(f"Butterworth filter frequencies invalid: low={low}, high={high}")
    b, a = butter(order, [low, high], btype='band')
    return b, a

def bandpass_filter(data, lowcut, highcut, fs, order=5):
    b, a = butter_bandpass(lowcut, highcut, fs, order=order)
    y = filtfilt(b, a, data, axis=-1)
    return y

LOWCUT = 20
HIGHCUT = 450
FILTER_ORDER = 4

with timer("Applying Signal Filtering"):
    X_train_val_filtered = np.zeros_like(X_train_val_raw)
    for i in range(N_SESSIONS_TRAIN_VAL):
        dc_removed = X_train_val_raw[i] - np.mean(X_train_val_raw[i], axis=-1, keepdims=True)
        X_train_val_filtered[i] = bandpass_filter(dc_removed, LOWCUT, HIGHCUT, FS, order=FILTER_ORDER)
    X_train_val_to_window = X_train_val_filtered
    print("Applied DC offset removal and bandpass filtering.")

del X_train_val_raw # Free memory
gc.collect()

# %% [markdown]
# ## 3. Dataset Preparation: Overlapping Windows (Identical to Foucauld's)

# %%
WINDOW_SIZE = 500
OVERLAP_PERCENTAGE = 0.75
step = int(WINDOW_SIZE * (1 - OVERLAP_PERCENTAGE))

def create_overlapping_windows(X_raw, y_raw, window_size, step):
    n_sessions, n_electrodes, n_timepoints_total = X_raw.shape
    _, n_joints, _ = y_raw.shape
    X_windowed_list, y_windowed_list, groups_list = [], [], []
    for session_idx in range(n_sessions):
        session_X, session_y = X_raw[session_idx], y_raw[session_idx]
        start_indices = np.arange(0, n_timepoints_total - window_size + 1, step)
        for start_idx in start_indices:
            end_idx = start_idx + window_size
            X_windowed_list.append(session_X[:, start_idx:end_idx])
            y_windowed_list.append(session_y[:, end_idx - 1])
            groups_list.append(session_idx)
    return np.stack(X_windowed_list), np.stack(y_windowed_list), np.array(groups_list)

with timer("Creating overlapping windows"):
    X_windowed, y_windowed, groups = create_overlapping_windows(
        X_train_val_to_window, y_train_val_raw, WINDOW_SIZE, step
    )
print(f"X_windowed: {X_windowed.shape}, y_windowed: {y_windowed.shape}, Groups: {groups.shape}")

del X_train_val_to_window, y_train_val_raw # Free memory
gc.collect()

# %% [markdown]
# ## 4. Custom Feature Extractor (Identical to Foucauld's `EMGFeatureExtractor`)

# %%
class EMGFeatureExtractor(BaseEstimator, TransformerMixin):
    def __init__(self, zc_threshold=1e-5, ssc_threshold=1e-5, selected_features=None, fs=FS, verbose=False):
        self.zc_threshold = zc_threshold
        self.ssc_threshold = ssc_threshold
        self.selected_features = selected_features
        self.fs = fs
        self.verbose = verbose
        self.all_available_feature_names = sorted(['MAV', 'RMS', 'Var', 'STD', 'ZC', 'MPR', 'WL', 'SSC', 'IEMG', 'MNF', 'MDF'])
        self._actual_features_to_extract = []
        self.n_electrodes_ = None

    def _resolve_features_to_extract(self):
        if self.selected_features is None: return self.all_available_feature_names
        else:
            validated_selection = [f for f in list(self.selected_features) if f in self.all_available_feature_names]
            if not validated_selection: raise ValueError("No valid features selected.")
            return sorted(list(set(validated_selection)))

    def fit(self, X, y=None):
        if X.ndim != 3: raise ValueError(f"Input X must be 3D, got shape {X.shape}")
        self.n_electrodes_ = X.shape[1]
        self._actual_features_to_extract = self._resolve_features_to_extract()
        if self.verbose: print(f"Feature Extractor fit. Electrodes: {self.n_electrodes_}. Extracting: {self._actual_features_to_extract}")
        return self

    def transform(self, X):
        if self.n_electrodes_ is None: self.fit(X)
        n_windows, n_electrodes, window_size = X.shape
        if n_electrodes != self.n_electrodes_: raise ValueError(f"Input X has {n_electrodes} electrodes, fitted on {self.n_electrodes_}")
        n_selected = len(self._actual_features_to_extract)
        if n_selected == 0: return np.empty((n_windows, 0))
        
        features_data = np.zeros((n_windows, self.n_electrodes_ * n_selected))
        freqs = np.fft.rfftfreq(window_size, d=1.0/self.fs)

        for i in range(n_windows):
            window_feats = []
            for j in range(self.n_electrodes_):
                ch_data = X[i, j, :]
                s_ch_feats = {}
                abs_ch_data = np.abs(ch_data)
                s_ch_feats['MAV'] = np.mean(abs_ch_data)
                s_ch_feats['RMS'] = np.sqrt(np.mean(ch_data**2))
                s_ch_feats['Var'] = np.var(ch_data)
                std_val = np.sqrt(max(0, s_ch_feats['Var']))
                s_ch_feats['STD'] = std_val
                s_ch_feats['ZC'] = np.sum(np.diff(np.signbit(ch_data - self.zc_threshold)).astype(int) != 0)
                s_ch_feats['MPR'] = np.mean(abs_ch_data > (std_val if std_val > 1e-9 else 1e-9))
                s_ch_feats['WL'] = np.sum(np.abs(np.diff(ch_data)))
                diff_d = np.diff(ch_data)
                s_ch_feats['SSC'] = np.sum(((diff_d[:-1] > self.ssc_threshold) & (diff_d[1:] < -self.ssc_threshold)) | ((diff_d[:-1] < -self.ssc_threshold) & (diff_d[1:] > self.ssc_threshold)))
                s_ch_feats['IEMG'] = np.sum(abs_ch_data)
                
                fft_spec = np.abs(np.fft.rfft(ch_data))
                pow_spec = fft_spec**2
                tot_pow = np.sum(pow_spec)
                if tot_pow > 1e-9:
                    s_ch_feats['MNF'] = np.sum(freqs * pow_spec) / tot_pow
                    cum_pow = np.cumsum(pow_spec)
                    mdf_idx = np.searchsorted(cum_pow, tot_pow / 2.0, side='left')
                    s_ch_feats['MDF'] = freqs[min(mdf_idx, len(freqs)-1)]
                else:
                    s_ch_feats['MNF'], s_ch_feats['MDF'] = 0.0, 0.0
                for feat_name in self._actual_features_to_extract: window_feats.append(s_ch_feats[feat_name])
            features_data[i, :] = window_feats
        return np.nan_to_num(features_data, nan=0.0, posinf=0.0, neginf=0.0)

    def get_feature_names_out(self, input_features=None):
        if not hasattr(self, '_actual_features_to_extract') or not self._actual_features_to_extract :
            temp_actual_features = self._resolve_features_to_extract()
            temp_n_electrodes = self.n_electrodes_ if self.n_electrodes_ is not None else N_ELECTRODES
        else:
            temp_actual_features = self._actual_features_to_extract
            temp_n_electrodes = self.n_electrodes_
        return [f"Elec{j+1}_{feat}" for j in range(temp_n_electrodes) for feat in temp_actual_features]

# Create canonical feature set (all features) for NNs and GBMs
with timer("Creating Canonical Feature Set (X_input_for_adv_models)"):
    feature_extractor_canonical = EMGFeatureExtractor(fs=FS)
    X_input_for_adv_models = feature_extractor_canonical.fit_transform(X_windowed)
    feature_names_canonical = feature_extractor_canonical.get_feature_names_out()
    print(f"X_input_for_adv_models shape: {X_input_for_adv_models.shape}")
    if np.any(np.isnan(X_input_for_adv_models)) or np.any(np.isinf(X_input_for_adv_models)):
        print("\nWARNING: Found NaNs or Infs in the canonical feature matrix!")

# %% [markdown]
# ## 5. Cross-Validation Strategy & Metrics (LOSO CV)

# %%
N_SPLITS_MAIN_CV = N_SESSIONS_TRAIN_VAL # = 5 for this dataset
cv_strategy_main = GroupKFold(n_splits=N_SPLITS_MAIN_CV)
print(f"Using GroupKFold with {N_SPLITS_MAIN_CV} splits (Leave-One-Session-Out) for main evaluation.")

# CV strategy for HPO (can be fewer folds for speed)
cv_strategy_hpo = GroupKFold(n_splits=N_FOLDS_HPO_CV)
print(f"Using GroupKFold with {N_FOLDS_HPO_CV} splits for HPO.")

# Performance Metrics (RMSE, NMSE, R2)
def root_mean_squared_error(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

def normalized_mean_squared_error(y_true, y_pred):
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    if y_true.ndim == 1: y_true, y_pred = y_true.reshape(-1, 1), y_pred.reshape(-1, 1)
    numerator = np.sum((y_true - y_pred)**2, axis=0) # Sum over samples for each joint
    mean_y_true_per_joint = np.mean(y_true, axis=0)
    denominator = np.sum((y_true - mean_y_true_per_joint)**2, axis=0) # Sum over samples
    
    # Handle cases where denominator is zero (all true values for a joint are the same)
    # Take mean over joints AFTER calculating NMSE per joint
    nmse_per_joint = np.zeros_like(denominator, dtype=float)
    valid_den = denominator > 1e-9
    nmse_per_joint[valid_den] = numerator[valid_den] / denominator[valid_den]
    nmse_per_joint[~valid_den] = 0.0 if np.all(numerator[~valid_den] < 1e-9) else np.inf
    return np.mean(nmse_per_joint) # Average NMSE across all joints

scorers = {
    'rmse': make_scorer(root_mean_squared_error, greater_is_better=False),
    'nmse': make_scorer(normalized_mean_squared_error, greater_is_better=False),
    'r2': make_scorer(r2_score, multioutput='variance_weighted') # More robust R2 for multi-target
}

all_cv_results_summary_list = [] # Store summary dicts for final DataFrame

# %% [markdown]
# ## 6. Baseline Models (Ridge, Lasso, RandomForest)

# %%
model_ridge = Ridge(alpha=1.0, random_state=RANDOM_STATE)
model_lasso = Lasso(alpha=0.01, max_iter=3000, tol=1e-3, random_state=RANDOM_STATE)
model_rf = RandomForestRegressor(
    n_estimators=100, max_depth=15, min_samples_split=10, min_samples_leaf=5,
    max_features=0.33, n_jobs=N_JOBS, random_state=RANDOM_STATE
)
baseline_models_sklearn = {"Ridge": model_ridge, "Lasso": model_lasso, "RandomForest": model_rf}

print("--- Evaluating Scikit-learn Baseline Models (LOSO CV) ---")
for model_name, model in baseline_models_sklearn.items():
    with timer(f"Cross-validating {model_name}"):
        pipeline = Pipeline([
            ('feature_extraction', EMGFeatureExtractor(fs=FS)), # Uses all features
            ('scaling', StandardScaler()),
            ('regression', clone(model))
        ])
        cv_args = {'estimator': pipeline, 'X': X_windowed, 'y': y_windowed,
                     'cv': cv_strategy_main, 'scoring': scorers, 'n_jobs': N_JOBS,
                     'return_train_score': False, 'error_score': 'raise'}
        if get_config().get('enable_metadata_routing', False): cv_args['params'] = {'groups': groups}
        else: cv_args['groups'] = groups
        
        try:
            scores_dict = cross_validate(**cv_args)
        except TypeError: # Fallback if groups not handled by params
            del cv_args['params']; cv_args['groups'] = groups
            scores_dict = cross_validate(**cv_args)

        mean_rmse = -np.mean(scores_dict['test_rmse']); std_rmse = np.std(scores_dict['test_rmse'])
        mean_nmse = -np.mean(scores_dict['test_nmse']); std_nmse = np.std(scores_dict['test_nmse'])
        mean_r2 = np.mean(scores_dict['test_r2']); std_r2 = np.std(scores_dict['test_r2'])
        fit_time = np.mean(scores_dict['fit_time'])
        
        print(f"  {model_name} CV Results:")
        print(f"    RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}")
        print(f"    NMSE: {mean_nmse:.4f} +/- {std_nmse:.4f}")
        print(f"    R2:   {mean_r2:.4f} +/- {std_r2:.4f}")
        all_cv_results_summary_list.append({
            'Model': model_name, 'Type': 'Baseline_Sklearn',
            'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse,
            'NMSE_mean': mean_nmse, 'NMSE_std': std_nmse,
            'R2_mean': mean_r2, 'R2_std': std_r2, 'FitTime': fit_time
        })

# %% [markdown]
# ## 7. Advanced Models: Gradient Boosting (XGBoost, LightGBM)
# These will use the `X_input_for_adv_models` (all features, pre-extracted).

# %%
# For GBM HPO, we need a wrapper function for Optuna
def gbm_objective(trial, X, y, groups, cv_strategy, model_type):
    if model_type == 'xgb':
        params = {
            'objective': 'reg:squarederror', 'eval_metric': 'rmse', 'n_estimators': trial.suggest_int('n_estimators', 100, 500),
            'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
            'max_depth': trial.suggest_int('max_depth', 3, 10), 'subsample': trial.suggest_float('subsample', 0.6, 1.0),
            'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
            'gamma': trial.suggest_float('gamma', 0, 0.5), 'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 1.0, log=True),
            'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 1.0, log=True),
            'random_state': RANDOM_STATE, 'n_jobs': N_JOBS,
            'tree_method': 'hist' if device == torch.device("cpu") else 'gpu_hist', # Use GPU if available
             'predictor': 'gpu_predictor' if device != torch.device("cpu") else 'auto',
        }
        model = xgb.XGBRegressor(**params)
    elif model_type == 'lgb':
        params = {
            'objective': 'regression_l1', 'metric': 'rmse', 'n_estimators': trial.suggest_int('n_estimators', 100, 500),
            'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
            'num_leaves': trial.suggest_int('num_leaves', 20, 150), 'max_depth': trial.suggest_int('max_depth', -1, 10), # -1 for no limit
            'subsample': trial.suggest_float('subsample', 0.6, 1.0), 'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
            'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 1.0, log=True), 'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 1.0, log=True),
            'min_child_samples': trial.suggest_int('min_child_samples', 5, 50),
            'random_state': RANDOM_STATE, 'n_jobs': N_JOBS,
            'device_type': 'gpu' if device != torch.device("cpu") else 'cpu',
            'force_col_wise': True, # Recommended for LightGBM for cache efficiency
             'verbose': -1
        }
        model = lgb.LGBMRegressor(**params)
    else:
        raise ValueError("Unsupported model_type for GBM objective.")

    pipeline = Pipeline([('scaling', StandardScaler()), ('regression', model)])
    cv_args = {'estimator': pipeline, 'X': X, 'y': y, 'cv': cv_strategy,
                 'scoring': scorers['rmse'], 'n_jobs': 1} # Optuna handles parallel trials
    if get_config().get('enable_metadata_routing', False): cv_args['params'] = {'groups': groups}
    else: cv_args['groups'] = groups
    
    try:
        scores = cross_validate(**cv_args)['test_score']
    except TypeError:
        del cv_args['params']; cv_args['groups'] = groups
        scores = cross_validate(**cv_args)['test_score']
        
    return -np.mean(scores) # Optuna minimizes, so return negative RMSE (which is positive)

advanced_models_gbm = {}
gbm_best_params = {}

for gbm_name, gbm_type_flag in [("XGBoost", "xgb"), ("LightGBM", "lgb")]:
    with timer(f"Hyperparameter Optimization for {gbm_name} using Optuna ({N_OPTUNA_TRIALS_GBM} trials)"):
        study = optuna.create_study(direction='minimize', pruner=optuna.pruners.MedianPruner(n_warmup_steps=3))
        study.optimize(lambda trial: gbm_objective(trial, X_input_for_adv_models, y_windowed, groups, cv_strategy_hpo, gbm_type_flag),
                       n_trials=N_OPTUNA_TRIALS_GBM, n_jobs=1) # n_jobs=1 for Optuna, it parallelizes trials
        gbm_best_params[gbm_name] = study.best_params
        print(f"Best params for {gbm_name}: {study.best_params}")
        print(f"Best RMSE from HPO for {gbm_name}: {-study.best_value:.4f}")

        if gbm_type_flag == 'xgb':
            # Add tree_method and predictor back if they were part of the original definition
            final_xgb_params = {**study.best_params, 'random_state': RANDOM_STATE, 'n_jobs': N_JOBS,
                                'tree_method': 'hist' if device == torch.device("cpu") else 'gpu_hist',
                                'predictor': 'gpu_predictor' if device != torch.device("cpu") else 'auto'}
            advanced_models_gbm[gbm_name] = xgb.XGBRegressor(**final_xgb_params)
        elif gbm_type_flag == 'lgb':
            final_lgb_params = {**study.best_params, 'random_state': RANDOM_STATE, 'n_jobs': N_JOBS,
                                'device_type': 'gpu' if device != torch.device("cpu") else 'cpu',
                                 'verbose': -1, 'force_col_wise':True}
            advanced_models_gbm[gbm_name] = lgb.LGBMRegressor(**final_lgb_params)


print("\n--- Evaluating Tuned GBMs (LOSO CV) ---")
for model_name, model in advanced_models_gbm.items():
    with timer(f"Cross-validating {model_name}"):
        pipeline = Pipeline([('scaling', StandardScaler()), ('regression', clone(model))])
        # GBMs directly use the pre-extracted features X_input_for_adv_models
        cv_args = {'estimator': pipeline, 'X': X_input_for_adv_models, 'y': y_windowed,
                     'cv': cv_strategy_main, 'scoring': scorers, 'n_jobs': N_JOBS,
                     'return_train_score': False, 'error_score': 'raise'}
        if get_config().get('enable_metadata_routing', False): cv_args['params'] = {'groups': groups}
        else: cv_args['groups'] = groups

        try:
            scores_dict = cross_validate(**cv_args)
        except TypeError:
            del cv_args['params']; cv_args['groups'] = groups
            scores_dict = cross_validate(**cv_args)

        mean_rmse = -np.mean(scores_dict['test_rmse']); std_rmse = np.std(scores_dict['test_rmse'])
        mean_nmse = -np.mean(scores_dict['test_nmse']); std_nmse = np.std(scores_dict['test_nmse'])
        mean_r2 = np.mean(scores_dict['test_r2']); std_r2 = np.std(scores_dict['test_r2'])
        fit_time = np.mean(scores_dict['fit_time'])

        print(f"  {model_name} CV Results:")
        print(f"    RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}")
        print(f"    NMSE: {mean_nmse:.4f} +/- {std_nmse:.4f}")
        print(f"    R2:   {mean_r2:.4f} +/- {std_r2:.4f}")
        all_cv_results_summary_list.append({
            'Model': model_name, 'Type': 'Advanced_GBM',
            'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse,
            'NMSE_mean': mean_nmse, 'NMSE_std': std_nmse,
            'R2_mean': mean_r2, 'R2_std': std_r2, 'FitTime': fit_time
        })
gc.collect()

# %% [markdown]
# ## 8. Feature Importance and Selection (SFS)
# SFS will use Ridge as the estimator and the canonical feature set.

# %%
with timer("Sequential Feature Selection (SFS) using Ridge"):
    sfs_estimator = Ridge(alpha=1.0, random_state=RANDOM_STATE)
    # SFS pipeline: Scaling -> Ridge (operates on X_input_for_adv_models)
    sfs_pipeline_internal = Pipeline([('scaling', StandardScaler()), ('regression', sfs_estimator)])
    
    sfs = SequentialFeatureSelector(
        estimator=sfs_pipeline_internal, n_features_to_select='auto', tol=1e-4, direction='forward',
        scoring=scorers['rmse'], cv=cv_strategy_main, n_jobs=N_JOBS
    )
    sfs_args = {'X': X_input_for_adv_models, 'y': y_windowed}
    if get_config().get('enable_metadata_routing', False): sfs_args['groups'] = groups
    else: pass # SFS might not directly use groups if estimator doesn't, but CV inside does

    sfs.fit(**sfs_args)
    selected_feature_indices_sfs = sfs.get_support(indices=True)
    selected_feature_names_sfs = [feature_names_canonical[i] for i in selected_feature_indices_sfs]
    n_selected_features_sfs = len(selected_feature_names_sfs)
    print(f"SFS selected {n_selected_features_sfs} features out of {len(feature_names_canonical)}.")
    
    # Store unique feature *types* selected by SFS
    selected_feature_types_sfs = []
    if n_selected_features_sfs > 0:
        try:
            selected_feature_types_sfs = sorted(list(set(f.split('_',1)[1] for f in selected_feature_names_sfs)))
            print(f"SFS selected feature types: {selected_feature_types_sfs}")
        except IndexError:
            print("Warning: Could not parse feature types from SFS names.")

# Evaluate best sklearn baseline with SFS features
if n_selected_features_sfs > 0 and n_selected_features_sfs < len(feature_names_canonical):
    # Find best sklearn baseline model from previous results
    df_temp_summary = pd.DataFrame(all_cv_results_summary_list)
    best_sklearn_baseline_name = df_temp_summary[df_temp_summary['Type'] == 'Baseline_Sklearn'].sort_values('RMSE_mean').iloc[0]['Model']
    best_sklearn_baseline_model = baseline_models_sklearn[best_sklearn_baseline_name]
    
    model_name_sfs_eval = f"{best_sklearn_baseline_name}_SFS"
    with timer(f"Cross-validating {model_name_sfs_eval}"):
        pipeline_sfs_eval = Pipeline([
            ('feature_extraction', EMGFeatureExtractor(fs=FS, selected_features=selected_feature_types_sfs)), # Extracts only SFS types
            ('scaling', StandardScaler()),
            ('regression', clone(best_sklearn_baseline_model))
        ])
        cv_args_sfs_eval = {'estimator': pipeline_sfs_eval, 'X': X_windowed, 'y': y_windowed,
                            'cv': cv_strategy_main, 'scoring': scorers, 'n_jobs': N_JOBS,
                            'return_train_score': False, 'error_score': 'raise'}
        if get_config().get('enable_metadata_routing', False): cv_args_sfs_eval['params'] = {'groups': groups}
        else: cv_args_sfs_eval['groups'] = groups
        
        try:
            scores_dict_sfs = cross_validate(**cv_args_sfs_eval)
        except TypeError:
            del cv_args_sfs_eval['params']; cv_args_sfs_eval['groups'] = groups
            scores_dict_sfs = cross_validate(**cv_args_sfs_eval)
            
        mean_rmse = -np.mean(scores_dict_sfs['test_rmse']); std_rmse = np.std(scores_dict_sfs['test_rmse'])
        mean_nmse = -np.mean(scores_dict_sfs['test_nmse']); std_nmse = np.std(scores_dict_sfs['test_nmse'])
        mean_r2 = np.mean(scores_dict_sfs['test_r2']); std_r2 = np.std(scores_dict_sfs['test_r2'])
        fit_time = np.mean(scores_dict_sfs['fit_time'])
        
        print(f"  {model_name_sfs_eval} CV Results:")
        print(f"    RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}")
        all_cv_results_summary_list.append({
            'Model': model_name_sfs_eval, 'Type': 'Baseline_SFS',
            'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse,
            'NMSE_mean': mean_nmse, 'NMSE_std': std_nmse,
            'R2_mean': mean_r2, 'R2_std': std_r2, 'FitTime': fit_time
        })
else:
    print("Skipping SFS model evaluation as SFS selected 0 or all features.")
gc.collect()

# %% [markdown]
# ## 9. Sophisticated Approach: Neural Network (NN_v3) with HPO

# %%
# --- PyTorch Helper Functions (DataPrefetcher, EarlyStopping, NN Model) ---
class DataPrefetcher:
    def __init__(self, loader, device):
        self.loader = iter(loader)
        self.device = device
        self.stream = torch.cuda.Stream()
        self.preload()
    def preload(self):
        try: self.next_input, self.next_target = next(self.loader)
        except StopIteration: self.next_input, self.next_target = None, None; return
        with torch.cuda.stream(self.stream):
            self.next_input = self.next_input.to(self.device, non_blocking=True).float()
            self.next_target = self.next_target.to(self.device, non_blocking=True).float()
    def next(self):
        torch.cuda.current_stream().wait_stream(self.stream)
        input_batch, target_batch = self.next_input, self.next_target
        self.preload()
        return input_batch, target_batch

class EarlyStop:
    def __init__(self, patience=10, min_delta=1e-5):
        self.patience, self.min_delta = patience, min_delta
        self.counter, self.best_loss, self.early_stop = 0, float('inf'), False
    def __call__(self, val_loss):
        if val_loss < self.best_loss - self.min_delta:
            self.best_loss, self.counter = val_loss, 0
        else:
            self.counter += 1
            if self.counter >= self.patience: self.early_stop = True
        return self.early_stop

class EMG_NN(nn.Module):
    def __init__(self, n_features, n_joints, hidden_layers_config, dropout_p):
        super().__init__()
        layers = []
        in_features = n_features
        for hidden_units in hidden_layers_config:
            layers.append(nn.Linear(in_features, hidden_units))
            layers.append(nn.BatchNorm1d(hidden_units))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout_p))
            in_features = hidden_units
        layers.append(nn.Linear(in_features, n_joints))
        self.network = nn.Sequential(*layers)
    def forward(self, x): return self.network(x)

def create_optimized_dataloader(X_data, y_data, batch_size, shuffle=True):
    dataset = TensorDataset(torch.tensor(X_data, dtype=torch.float32), torch.tensor(y_data, dtype=torch.float32))
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle, num_workers=NUM_WORKERS_LOADER, pin_memory=True, persistent_workers=(NUM_WORKERS_LOADER > 0))

def train_nn_epoch_amp(model, loader, criterion, optimizer, scaler, device):
    model.train(); running_loss = 0.0
    prefetcher = DataPrefetcher(loader, device)
    inputs, targets = prefetcher.next()
    while inputs is not None:
        optimizer.zero_grad(set_to_none=True)
        with autocast(enabled=(device.type == 'cuda')): # autocast only if cuda
            outputs = model(inputs)
            loss = criterion(outputs, targets)
        if device.type == 'cuda':
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else: # For CPU
            loss.backward()
            optimizer.step()
        running_loss += loss.item() * inputs.size(0)
        inputs, targets = prefetcher.next()
    return running_loss / len(loader.dataset)

def evaluate_nn_amp(model, loader, criterion, device):
    model.eval(); running_loss = 0.0
    all_preds_list, all_targets_list = [], []
    prefetcher = DataPrefetcher(loader, device)
    inputs, targets = prefetcher.next()
    with torch.no_grad():
        while inputs is not None:
            with autocast(enabled=(device.type == 'cuda')):
                outputs = model(inputs)
                loss = criterion(outputs, targets)
            running_loss += loss.item() * inputs.size(0)
            all_preds_list.append(outputs.cpu().numpy())
            all_targets_list.append(targets.cpu().numpy())
            inputs, targets = prefetcher.next()
    final_mse = running_loss / len(loader.dataset)
    return final_mse, np.concatenate(all_preds_list), np.concatenate(all_targets_list)

# --- Optuna Objective Function for NN HPO ---
def nn_objective(trial, X_features, y_targets, groups_all, cv_hpo_strategy):
    # Hyperparameters to tune
    n_layers = trial.suggest_int('n_layers', 2, 4)
    hidden_config = [trial.suggest_int(f'n_units_l{i}', 64, 512, log=True) for i in range(n_layers)]
    dropout_p = trial.suggest_float('dropout_p', 0.1, 0.5)
    lr = trial.suggest_float('lr', 1e-4, 1e-2, log=True)
    weight_decay = trial.suggest_float('weight_decay', 1e-6, 1e-3, log=True)
    batch_size = trial.suggest_categorical('batch_size', [128, 256, 512, 1024])
    optimizer_name = trial.suggest_categorical('optimizer', ['Adam', 'AdamW'])

    fold_val_rmses = []
    for fold_idx, (train_idx, val_idx) in enumerate(cv_hpo_strategy.split(X_features, y_targets, groups=groups_all)):
        X_train_fold, X_val_fold = X_features[train_idx], X_features[val_idx]
        y_train_fold, y_val_fold = y_targets[train_idx], y_targets[val_idx]

        scaler_fold = StandardScaler().fit(X_train_fold)
        X_train_fold_scaled, X_val_fold_scaled = scaler_fold.transform(X_train_fold), scaler_fold.transform(X_val_fold)

        train_loader = create_optimized_dataloader(X_train_fold_scaled, y_train_fold, batch_size, shuffle=True)
        val_loader = create_optimized_dataloader(X_val_fold_scaled, y_val_fold, batch_size * 2, shuffle=False)

        model = EMG_NN(X_features.shape[1], N_JOINTS, hidden_config, dropout_p).to(device)
        criterion = nn.MSELoss()
        if optimizer_name == 'Adam': optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
        else: optimizer = optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)
        
        amp_scaler = GradScaler(enabled=(device.type == 'cuda'))
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=PATIENCE_LR_SCHEDULER//2, factor=0.5, verbose=False)
        early_stopper = EarlyStop(patience=PATIENCE_EARLY_STOPPING//2)

        for epoch in range(MAX_EPOCHS_NN_V3 // 2): # Fewer epochs for HPO
            train_nn_epoch_amp(model, train_loader, criterion, optimizer, amp_scaler, device)
            mse_val, _, _ = evaluate_nn_amp(model, val_loader, criterion, device)
            scheduler.step(mse_val)
            if early_stopper(mse_val): break
        
        final_mse_val, _, _ = evaluate_nn_amp(model, val_loader, criterion, device)
        fold_val_rmses.append(np.sqrt(final_mse_val))
        
        # Optuna Pruning
        trial.report(np.mean(fold_val_rmses), fold_idx) # Report intermediate value per fold
        if trial.should_prune(): raise optuna.TrialPruned()
            
    mean_rmse_hpo = np.mean(fold_val_rmses)
    # Clean up
    del model, train_loader, val_loader, optimizer, scheduler, early_stopper
    if device.type == 'cuda': torch.cuda.empty_cache()
    gc.collect()
    return mean_rmse_hpo

# --- Run NN HPO ---
best_nn_params = None
with timer(f"Hyperparameter Optimization for NN_v3 using Optuna ({N_OPTUNA_TRIALS_NN} trials)"):
    study_nn = optuna.create_study(direction='minimize', pruner=optuna.pruners.MedianPruner(n_warmup_steps=N_FOLDS_HPO_CV * 3)) # Prune after 3 full CV passes
    study_nn.optimize(lambda trial: nn_objective(trial, X_input_for_adv_models, y_windowed, groups, cv_strategy_hpo),
                      n_trials=N_OPTUNA_TRIALS_NN, n_jobs=1, gc_after_trial=True) # n_jobs=1 for Optuna with PyTorch GPU
    best_nn_params = study_nn.best_params
    print(f"Best NN_v3 params from HPO: {best_nn_params}")
    print(f"Best NN_v3 RMSE from HPO: {study_nn.best_value:.4f}")

# --- Evaluate Tuned NN_v3 with Full Main CV ---
print("\n--- Evaluating Tuned NN_v3 (LOSO CV) ---")
fold_rmses_nn, fold_nmses_nn, fold_r2s_nn, fold_fit_times_nn = [], [], [], []

for k_fold, (train_idx, val_idx) in enumerate(cv_strategy_main.split(X_input_for_adv_models, y_windowed, groups=groups)):
    fold_start_time = time.time()
    print(f"  Processing NN_v3 Main CV Fold {k_fold+1}/{N_SPLITS_MAIN_CV}")
    X_train_f, X_val_f = X_input_for_adv_models[train_idx], X_input_for_adv_models[val_idx]
    y_train_f, y_val_f = y_windowed[train_idx], y_windowed[val_idx]

    scaler_f = StandardScaler().fit(X_train_f)
    X_train_f_scaled, X_val_f_scaled = scaler_f.transform(X_train_f), scaler_f.transform(X_val_f)
    train_loader_f = create_optimized_dataloader(X_train_f_scaled, y_train_f, best_nn_params['batch_size'], shuffle=True)
    val_loader_f = create_optimized_dataloader(X_val_f_scaled, y_val_f, best_nn_params['batch_size'] * 2, shuffle=False)

    model_f = EMG_NN(X_input_for_adv_models.shape[1], N_JOINTS,
                     [best_nn_params[f'n_units_l{i}'] for i in range(best_nn_params['n_layers'])],
                     best_nn_params['dropout_p']).to(device)
    criterion_f = nn.MSELoss()
    opt_name = best_nn_params.get('optimizer', 'AdamW') # Default to AdamW if not tuned
    if opt_name == 'Adam': optimizer_f = optim.Adam(model_f.parameters(), lr=best_nn_params['lr'], weight_decay=best_nn_params['weight_decay'])
    else: optimizer_f = optim.AdamW(model_f.parameters(), lr=best_nn_params['lr'], weight_decay=best_nn_params['weight_decay'])

    amp_scaler_f = GradScaler(enabled=(device.type == 'cuda'))
    scheduler_f = optim.lr_scheduler.ReduceLROnPlateau(optimizer_f, 'min', patience=PATIENCE_LR_SCHEDULER, factor=0.5, verbose=False)
    early_stopper_f = EarlyStop(patience=PATIENCE_EARLY_STOPPING)

    for epoch_f in range(MAX_EPOCHS_NN_V3):
        train_nn_epoch_amp(model_f, train_loader_f, criterion_f, optimizer_f, amp_scaler_f, device)
        mse_val_f, _, _ = evaluate_nn_amp(model_f, val_loader_f, criterion_f, device)
        scheduler_f.step(mse_val_f)
        if early_stopper_f(mse_val_f): print(f"    Fold {k_fold+1} early stopping at epoch {epoch_f+1}."); break
    
    final_mse_f, y_pred_f, y_true_f = evaluate_nn_amp(model_f, val_loader_f, criterion_f, device)
    fold_fit_times_nn.append(time.time() - fold_start_time)
    fold_rmses_nn.append(np.sqrt(final_mse_f))
    fold_nmses_nn.append(normalized_mean_squared_error(y_true_f, y_pred_f))
    fold_r2s_nn.append(r2_score(y_true_f, y_pred_f, multioutput='variance_weighted'))
    print(f"    Fold {k_fold+1} Val RMSE: {fold_rmses_nn[-1]:.4f}, NMSE: {fold_nmses_nn[-1]:.4f}, R2: {fold_r2s_nn[-1]:.4f}")
    del model_f, train_loader_f, val_loader_f, optimizer_f, scheduler_f, early_stopper_f
    if device.type == 'cuda': torch.cuda.empty_cache(); gc.collect()

# Store NN_v3 results
mean_rmse_nn, std_rmse_nn = np.mean(fold_rmses_nn), np.std(fold_rmses_nn)
mean_nmse_nn, std_nmse_nn = np.mean(fold_nmses_nn), np.std(fold_nmses_nn)
mean_r2_nn, std_r2_nn = np.mean(fold_r2s_nn), np.std(fold_r2s_nn)
fit_time_nn = np.mean(fold_fit_times_nn)
print(f"\nNN_v3 Tuned CV Summary: RMSE: {mean_rmse_nn:.4f} +/- {std_rmse_nn:.4f}, R2: {mean_r2_nn:.4f} +/- {std_r2_nn:.4f}")
all_cv_results_summary_list.append({
    'Model': 'NeuralNetwork_v3_Tuned', 'Type': 'Sophisticated_NN',
    'RMSE_mean': mean_rmse_nn, 'RMSE_std': std_rmse_nn,
    'NMSE_mean': mean_nmse_nn, 'NMSE_std': std_nmse_nn,
    'R2_mean': mean_r2_nn, 'R2_std': std_r2_nn, 'FitTime': fit_time_nn
})
# Train final NN on all data (for ensembling and potential direct use)
with timer("Training Final NN_v3_Tuned on all data"):
    final_nn_scaler = StandardScaler().fit(X_input_for_adv_models)
    X_all_scaled_nn = final_nn_scaler.transform(X_input_for_adv_models)
    final_train_loader_nn = create_optimized_dataloader(X_all_scaled_nn, y_windowed, best_nn_params['batch_size'], shuffle=True)
    
    final_model_nn = EMG_NN(X_input_for_adv_models.shape[1], N_JOINTS,
                            [best_nn_params[f'n_units_l{i}'] for i in range(best_nn_params['n_layers'])],
                            best_nn_params['dropout_p']).to(device)
    final_criterion_nn = nn.MSELoss()
    opt_name = best_nn_params.get('optimizer', 'AdamW')
    if opt_name == 'Adam': final_optimizer_nn = optim.Adam(final_model_nn.parameters(), lr=best_nn_params['lr'], weight_decay=best_nn_params['weight_decay'])
    else: final_optimizer_nn = optim.AdamW(final_model_nn.parameters(), lr=best_nn_params['lr'], weight_decay=best_nn_params['weight_decay'])
    
    final_amp_scaler_nn = GradScaler(enabled=(device.type == 'cuda'))
    # Typically train for slightly more epochs or until training loss plateaus on full data
    num_final_epochs = int(MAX_EPOCHS_NN_V3 * 1.2) # e.g., 20% more epochs
    for epoch in range(num_final_epochs):
        loss = train_nn_epoch_amp(final_model_nn, final_train_loader_nn, final_criterion_nn, final_optimizer_nn, final_amp_scaler_nn, device)
        if (epoch + 1) % (num_final_epochs // 5) == 0: print(f"  Final NN Train Epoch {epoch+1}/{num_final_epochs}, Loss: {loss:.4f}")

    final_trained_model_objects = {'NeuralNetwork_v3_Tuned': {'model': final_model_nn, 'scaler': final_nn_scaler, 'feature_extractor': feature_extractor_canonical, 'model_type': 'NeuralNetwork_v3_Tuned'}}

# %% [markdown]
# ## 10. Ensembling Strategies

# %%
# --- Prepare All Base Models (Sklearn, GBMs) Trained on Full Data for Ensembling ---
# NNs are handled by `final_trained_model_objects`
all_sklearn_base_model_names = list(baseline_models_sklearn.keys()) + list(advanced_models_gbm.keys())
for model_name in all_sklearn_base_model_names:
    if model_name in baseline_models_sklearn: # Sklearn baselines (Ridge, Lasso, RF)
        model_instance = baseline_models_sklearn[model_name]
        p_full = Pipeline([('feature_extraction', EMGFeatureExtractor(fs=FS)), ('scaling', StandardScaler()), ('regression', clone(model_instance))])
        with timer(f"Fitting final '{model_name}' on all data (for ensembling)"): p_full.fit(X_windowed, y_windowed)
        final_trained_model_objects[model_name] = p_full
    elif model_name in advanced_models_gbm: # GBMs
        model_instance = advanced_models_gbm[model_name]
        p_full_gbm = Pipeline([('scaling', StandardScaler()), ('regression', clone(model_instance))])
        with timer(f"Fitting final '{model_name}' on all data (for ensembling)"): p_full_gbm.fit(X_input_for_adv_models, y_windowed)
        final_trained_model_objects[model_name] = p_full_gbm

# --- Generate Out-of-Fold (OOF) Predictions for All Base Models ---
oof_predictions = {}
# Models to get OOF predictions for (Sklearn baselines, GBMs, Tuned NN)
models_for_oof = {**baseline_models_sklearn, **advanced_models_gbm, 'NeuralNetwork_v3_Tuned': 'NN_Special_Case'}

for model_name, model_obj_or_flag in models_for_oof.items():
    print(f"\n--- Generating OOF predictions for: {model_name} ---")
    if model_name == 'NeuralNetwork_v3_Tuned':
        # Manual OOF generation for NN
        nn_oof_preds_array = np.zeros_like(y_windowed)
        for k_fold, (train_idx, val_idx) in enumerate(cv_strategy_main.split(X_input_for_adv_models, y_windowed, groups=groups)):
            print(f"  NN OOF Fold {k_fold+1}/{N_SPLITS_MAIN_CV}")
            X_tr_oof, X_val_oof = X_input_for_adv_models[train_idx], X_input_for_adv_models[val_idx]
            y_tr_oof, y_val_oof = y_windowed[train_idx], y_windowed[val_idx]
            sc_oof = StandardScaler().fit(X_tr_oof)
            X_tr_sc_oof, X_val_sc_oof = sc_oof.transform(X_tr_oof), sc_oof.transform(X_val_oof)
            tr_ld_oof = create_optimized_dataloader(X_tr_sc_oof, y_tr_oof, best_nn_params['batch_size'], True)
            val_ld_oof = create_optimized_dataloader(X_val_sc_oof, y_val_oof, best_nn_params['batch_size']*2, False)

            m_oof = EMG_NN(X_input_for_adv_models.shape[1], N_JOINTS, [best_nn_params[f'n_units_l{i}'] for i in range(best_nn_params['n_layers'])], best_nn_params['dropout_p']).to(device)
            crit_oof = nn.MSELoss()
            opt_name_oof = best_nn_params.get('optimizer', 'AdamW')
            if opt_name_oof == 'Adam': opt_oof = optim.Adam(m_oof.parameters(), lr=best_nn_params['lr'], weight_decay=best_nn_params['weight_decay'])
            else: opt_oof = optim.AdamW(m_oof.parameters(), lr=best_nn_params['lr'], weight_decay=best_nn_params['weight_decay'])
            
            amp_sc_oof = GradScaler(enabled=(device.type=='cuda'))
            sched_oof = optim.lr_scheduler.ReduceLROnPlateau(opt_oof, 'min', patience=PATIENCE_LR_SCHEDULER, factor=0.5)
            es_oof = EarlyStop(patience=PATIENCE_EARLY_STOPPING)
            for ep_oof in range(MAX_EPOCHS_NN_V3):
                train_nn_epoch_amp(m_oof, tr_ld_oof, crit_oof, opt_oof, amp_sc_oof, device)
                mse_v_oof, _, _ = evaluate_nn_amp(m_oof, val_ld_oof, crit_oof, device)
                sched_oof.step(mse_v_oof)
                if es_oof(mse_v_oof): break
            _, y_pred_val_oof, _ = evaluate_nn_amp(m_oof, val_ld_oof, crit_oof, device)
            nn_oof_preds_array[val_idx] = y_pred_val_oof
            del m_oof, tr_ld_oof, val_ld_oof, opt_oof, sched_oof, es_oof; gc.collect(); torch.cuda.empty_cache()
        oof_predictions[model_name] = nn_oof_preds_array
    else: # Sklearn Pipelines (including GBMs)
        if model_name in baseline_models_sklearn: # Baselines use raw windowed data
            p_oof = Pipeline([('feature_extraction', EMGFeatureExtractor(fs=FS)), ('scaling', StandardScaler()), ('regression', clone(model_obj_or_flag))])
            X_oof_input = X_windowed
        else: # GBMs use pre-extracted features
            p_oof = Pipeline([('scaling', StandardScaler()), ('regression', clone(model_obj_or_flag))])
            X_oof_input = X_input_for_adv_models
        
        cv_args_oof = {'estimator': p_oof, 'X': X_oof_input, 'y': y_windowed, 'cv': cv_strategy_main,
                       'n_jobs': N_JOBS, 'method': 'predict', 'error_score':'raise'}
        if get_config().get('enable_metadata_routing', False): cv_args_oof['params'] = {'groups': groups}
        else: cv_args_oof['groups'] = groups
        try:
            oof_preds = cross_val_predict(**cv_args_oof)
        except TypeError:
            del cv_args_oof['params']; cv_args_oof['groups'] = groups
            oof_preds = cross_val_predict(**cv_args_oof)
        oof_predictions[model_name] = oof_preds
    print(f"  OOF predictions shape for {model_name}: {oof_predictions[model_name].shape}")

# --- Averaging Ensemble ---
with timer("Evaluating Averaging Ensemble"):
    model_names_for_avg = list(oof_predictions.keys())
    oof_preds_sum_avg = np.sum([oof_predictions[name] for name in model_names_for_avg], axis=0)
    oof_preds_avg = oof_preds_sum_avg / len(model_names_for_avg)
    avg_rmse = root_mean_squared_error(y_windowed, oof_preds_avg)
    avg_nmse = normalized_mean_squared_error(y_windowed, oof_preds_avg)
    avg_r2 = r2_score(y_windowed, oof_preds_avg, multioutput='variance_weighted')
    print(f"  Averaging Ensemble OOF RMSE: {avg_rmse:.4f}, NMSE: {avg_nmse:.4f}, R2: {avg_r2:.4f}")
    all_cv_results_summary_list.append({'Model': 'AverageEnsemble', 'Type': 'Ensemble', 'RMSE_mean': avg_rmse, 'RMSE_std': np.nan, 'NMSE_mean': avg_nmse, 'NMSE_std': np.nan, 'R2_mean': avg_r2, 'R2_std': np.nan, 'FitTime': np.nan})

# --- Stacking Ensemble ---
X_meta = np.hstack([oof_predictions[name] for name in model_names_for_avg]) # Use same models as averaging for stacking
print(f"X_meta shape for stacking: {X_meta.shape}")

# Meta-learners with automatic alpha tuning
meta_learners = {
    "Stacking_RidgeCV": RidgeCV(alphas=np.logspace(-4, 2, 100), store_cv_values=False), # cv inside RidgeCV
    "Stacking_LassoCV": LassoCV(alphas=np.logspace(-5, -1, 100), max_iter=5000, tol=1e-3, random_state=RANDOM_STATE, n_jobs=N_JOBS),
    "Stacking_RF_Meta": RandomForestRegressor(n_estimators=50, max_depth=7, n_jobs=N_JOBS, random_state=RANDOM_STATE)
}
for meta_name, meta_model in meta_learners.items():
    with timer(f"Cross-validating Meta-Learner: {meta_name}"):
        # Added scaler for meta-features
        stacking_pipeline_meta = Pipeline([('meta_scaling', StandardScaler()), ('meta_regression', clone(meta_model))])
        cv_args_meta = {'estimator': stacking_pipeline_meta, 'X': X_meta, 'y': y_windowed,
                        'cv': cv_strategy_main, 'scoring': scorers, 'n_jobs': N_JOBS,
                        'return_train_score': False, 'error_score': 'raise'}
        # For RidgeCV/LassoCV, groups are handled by their internal CV. For RF, pass groups.
        if 'RF' in meta_name: # Only pass groups for RF meta
            if get_config().get('enable_metadata_routing', False): cv_args_meta['params'] = {'groups': groups}
            else: cv_args_meta['groups'] = groups
        
        try:
            scores_meta = cross_validate(**cv_args_meta)
        except TypeError as e: # Fallback if groups kwarg causes issue for RF
            if 'RF' in meta_name and 'groups' in str(e).lower():
                 del cv_args_meta['params'] # if it exists
                 if 'groups' in cv_args_meta: del cv_args_meta['groups']
                 scores_meta = cross_validate(**cv_args_meta)
            else: raise e


        mean_rmse = -np.mean(scores_meta['test_rmse']); std_rmse = np.std(scores_meta['test_rmse'])
        mean_nmse = -np.mean(scores_meta['test_nmse']); std_nmse = np.std(scores_meta['test_nmse'])
        mean_r2 = np.mean(scores_meta['test_r2']); std_r2 = np.std(scores_meta['test_r2'])
        fit_time = np.mean(scores_meta['fit_time'])
        print(f"  {meta_name} OOF RMSE: {mean_rmse:.4f} +/- {std_rmse:.4f}, R2: {mean_r2:.4f} +/- {std_r2:.4f}")
        all_cv_results_summary_list.append({'Model': meta_name, 'Type': 'Ensemble_Stacking', 'RMSE_mean': mean_rmse, 'RMSE_std': std_rmse, 'NMSE_mean': mean_nmse, 'NMSE_std': std_nmse, 'R2_mean': mean_r2, 'R2_std': std_r2, 'FitTime': fit_time})
gc.collect()

# %% [markdown]
# ## 11. Final Model Selection and Test Set Prediction Preparation

# %%
results_df_final = pd.DataFrame(all_cv_results_summary_list)
print("\n--- Final Combined Model Comparison Table (sorted by RMSE) ---")
print(results_df_final.sort_values(by='RMSE_mean').to_string(index=False, float_format="%.4f"))

best_model_row = results_df_final.loc[results_df_final['RMSE_mean'].idxmin()]
best_overall_model_name = best_model_row['Model']
print(f"\nSelected Best Overall Model for Guided Gestures: '{best_overall_model_name}' (Mean CV RMSE: {best_model_row['RMSE_mean']:.4f})")

# --- Prepare the final model object for prediction ---
final_model_for_prediction_guided = None

if best_overall_model_name == 'NeuralNetwork_v3_Tuned':
    final_model_for_prediction_guided = final_trained_model_objects['NeuralNetwork_v3_Tuned']
    print(f"'{best_overall_model_name}' is ready (was trained on full data).")

elif best_overall_model_name in final_trained_model_objects: # Sklearn baselines or GBMs trained on full data
    final_model_for_prediction_guided = final_trained_model_objects[best_overall_model_name]
    print(f"'{best_overall_model_name}' is ready (was trained on full data).")
    # If it was an SFS model, reconstruct its pipeline with the specific SFS feature extractor
    if best_overall_model_name.endswith("_SFS"):
        base_sfs_name = best_overall_model_name.replace("_SFS","")
        final_model_for_prediction_guided = Pipeline([
            ('feature_extraction', EMGFeatureExtractor(fs=FS, selected_features=selected_feature_types_sfs)), # Uses SFS types
            ('scaling', StandardScaler()),
            ('regression', clone(baseline_models_sklearn[base_sfs_name])) # Use the original regressor
        ])
        with timer(f"Re-fitting SFS pipeline '{best_overall_model_name}' on all data"):
            final_model_for_prediction_guided.fit(X_windowed, y_windowed)


elif best_overall_model_name == 'AverageEnsemble':
    final_model_for_prediction_guided = {
        'model_type': 'AverageEnsemble',
        'base_models_trained': {name: final_trained_model_objects[name] for name in model_names_for_avg},
        'base_model_order': model_names_for_avg # From previous cell
    }
    print(f"'{best_overall_model_name}' prepared (uses fully trained base models).")

elif best_overall_model_name in meta_learners: # Stacking ensemble
    final_meta_learner_instance = clone(meta_learners[best_overall_model_name])
    final_stacking_pipeline_meta = Pipeline([('meta_scaling', StandardScaler()), ('meta_regression', final_meta_learner_instance)])
    with timer(f"Fitting final meta-learner '{best_overall_model_name}' on full X_meta"):
        final_stacking_pipeline_meta.fit(X_meta, y_windowed)
    final_model_for_prediction_guided = {
        'model_type': 'StackingEnsemble', 'name': best_overall_model_name,
        'meta_model_pipeline': final_stacking_pipeline_meta,
        'base_models_trained': {name: final_trained_model_objects[name] for name in model_names_for_avg}, # Assuming same base models for avg and stack
        'base_model_order': model_names_for_avg
    }
    print(f"Final Stacking Ensemble '{best_overall_model_name}' prepared.")
else:
    raise ValueError(f"Don't know how to prepare final model for '{best_overall_model_name}'.")

if final_model_for_prediction_guided is None:
    raise RuntimeError("Failed to prepare the final model.")

# %% [markdown]
# ## 12. Generate Test Set Predictions (Guided Gestures)
# This section would load `guided_testset_X.npy`, process it, and predict.
# The logic for prediction will depend on the type of `final_model_for_prediction_guided`.

# %%
print("\n--- Generating Predictions for Guided Test Set (Placeholder) ---")
# X_TEST_GUIDED_PATH = "guided_testset_X.npy"
# with timer("Loading Guided Test Data"):
#     X_test_guided_raw = np.load(X_TEST_GUIDED_PATH) # (sessions, windows_per_session, electrodes, window_size)
#     n_ts, n_twps, n_te, t_ws = X_test_guided_raw.shape
#     X_test_guided_reshaped = X_test_guided_raw.reshape(-1, n_te, t_ws) # (total_windows, electrodes, window_size)

# guided_test_predictions = None
# with timer(f"Predicting with final model '{best_overall_model_name}' on test data"):
#     model_pred_obj = final_model_for_prediction_guided
#     if isinstance(model_pred_obj, Pipeline): # Sklearn baseline, SFS, GBM pipeline
#         if model_pred_obj.steps[0][0] == 'feature_extraction': # Needs raw EMG windows
#             guided_test_predictions = model_pred_obj.predict(X_test_guided_reshaped)
#         else: # GBM pipeline, needs features
#             X_test_guided_features = feature_extractor_canonical.transform(X_test_guided_reshaped) # Use transform only
#             guided_test_predictions = model_pred_obj.predict(X_test_guided_features)
#     elif isinstance(model_pred_obj, dict) and model_pred_obj.get('model_type') == 'NeuralNetwork_v3_Tuned':
#         # NN prediction logic (extract, scale, predict)
#         nn_model = model_pred_obj['model']
#         nn_scaler = model_pred_obj['scaler']
#         nn_extractor = model_pred_obj['feature_extractor'] # This should be the canonical one
#         X_test_feat_nn = nn_extractor.transform(X_test_guided_reshaped) # Use transform only
#         X_test_feat_sc_nn = nn_scaler.transform(X_test_feat_nn)
#         test_ds_nn = TensorDataset(torch.tensor(X_test_feat_sc_nn, dtype=torch.float32))
#         test_ld_nn = DataLoader(test_ds_nn, batch_size=best_nn_params['batch_size']*4, shuffle=False, num_workers=NUM_WORKERS_LOADER, pin_memory=True)
#         nn_model.eval()
#         preds_list = []
#         prefetcher = DataPrefetcher(test_ld_nn, device) # TestDataPrefetcher needs 1 input
#         inputs, _ = prefetcher.next() # Ignore target if TestDataPrefetcher returns it
#         with torch.no_grad():
#             while inputs is not None:
#                 with autocast(enabled=(device.type=='cuda')): outputs = nn_model(inputs)
#                 preds_list.append(outputs.cpu().numpy())
#                 inputs, _ = prefetcher.next()
#         guided_test_predictions = np.concatenate(preds_list, axis=0)
#     elif isinstance(model_pred_obj, dict) and model_pred_obj.get('model_type') == 'AverageEnsemble':
#         # AverageEnsemble prediction logic
#         # ... (similar to OOF, but on test data and using fully trained base models) ...
#         pass # Placeholder
#     elif isinstance(model_pred_obj, dict) and model_pred_obj.get('model_type') == 'StackingEnsemble':
#         # StackingEnsemble prediction logic
#         # ... (similar to OOF, but on test data and using fully trained base models & meta-model) ...
#         pass # Placeholder
#     else:
#         raise TypeError(f"Unsupported final model type for test prediction: {type(model_pred_obj)}")

# if guided_test_predictions is not None:
#     print(f"Shape of final guided test predictions: {guided_test_predictions.shape}")
#     # Expected: (1660, 51)
# else:
#     print("Test predictions not generated (logic is placeholder).")

print("Placeholder for test set prediction. Implement based on chosen model type.")
print("The `final_model_for_prediction_guided` object holds the trained model/components.")
print("The `final_trained_model_objects` dictionary holds all fully trained base models if needed for ensembles.")

# %% [markdown]
# ---
# End of Enhanced Guided Gestures Script. The next steps would be to fully implement
# the test prediction logic for all model types and then integrate the freemoves dataset.
# ---
```
