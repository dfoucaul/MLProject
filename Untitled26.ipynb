```python
# ⚙️ Fonction pour extraire plusieurs features par fenêtre
def extract_features(window):
    features = []
    for channel in window:  # boucle sur les 8 électrodes
        rms = np.sqrt(np.mean(channel**2))
        mav = np.mean(np.abs(channel))
        std = np.std(channel)
        zc = np.sum(np.diff(np.sign(channel)) != 0)
        features += [rms, mav, std, zc]
    return features  # 4 features × 8 canaux = 32 features

# 🏗 Construction du dataset
X_train = []
y_train = []

for i in range(n_windows):
    start = i * step
    end = start + window_size

    X_win = X[:, start:end]
    y_win = y[:, start:end]

    features = extract_features(X_win)          # → shape: (32,)
    target = np.mean(y_win, axis=1).tolist()    # → shape: (51,)

    X_train.append(features)
    y_train.append(target)

X_train = np.array(X_train)  # shape (n_samples, 32)
y_train = np.array(y_train)  # shape (n_samples, 51)

print("Shape X_train :", X_train.shape)
print("Shape y_train :", y_train.shape)
```

```python
import numpy as np
from pyriemann.estimation import Covariances
from pyriemann.tangentspace import TangentSpace
from sklearn.linear_model import RidgeCV
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline

# Charger les données
X = np.load("guided_dataset_X.npy", mmap_mode="r")  # (5, 8, 230000)
y = np.load("guided_dataset_y.npy", mmap_mode="r")

window_size = 500
step = 250

X_cov = []
y_cov = []

# Fenêtrage & récupération des fenêtres (8, 500)
for session in range(5):
    X_session = X[session]
    y_session = y[session]
    n_windows = (X_session.shape[1] - window_size) // step + 1

    for i in range(n_windows):
        start = i * step
        end = start + window_size
        emg_win = X_session[:, start:end]  # (8, 500)
        pose_win = y_session[:, start:end]  # (51, 500)

        cov = np.cov(emg_win)  # (8, 8)
        X_cov.append(cov)
        y_cov.append(np.mean(pose_win, axis=1))  # Moyenne des 51 angles

X_cov = np.array(X_cov)  # (nb_samples, 8, 8)
y_cov = np.array(y_cov)  # (nb_samples, 51)

# Pipeline: covariance -> tangent space -> regression
pipe = Pipeline([
    ("ts", TangentSpace()),
    ("reg", MultiOutputRegressor(RidgeCV(alphas=[0.1, 1.0, 10.0])))
])

# Split train/val
X_train, X_val, y_train, y_val = train_test_split(X_cov, y_cov, test_size=0.2, random_state=42)

# Entraînement
pipe.fit(X_train, y_train)

# Prédiction + erreur
y_pred = pipe.predict(X_val)
mae = np.mean(np.abs(y_pred - y_val))
mean_true = np.mean(np.abs(y_val)) + 1e-8
rel_err = (mae / mean_true) * 100
print(f"📊 Erreur relative moyenne (covariance pipeline) : {rel_err:.2f}%")
```
